## Object-Oriented Programming (OOP)

### What is OOP?
Object-Oriented Programming (OOP) is a programming paradigm that organizes code using **classes and objects**.  
It helps in writing **modular, reusable, and maintainable code**.


### Why OOP?
- Helps in **data abstraction**
- Improves **code reusability**
- Makes code **organized and scalable**
- Models real-world entities


##  Class and Object

### Class
A **class** is a blueprint or template used to create objects.

###  Object
An **object** is an instance of a class. It contains **data (attributes)** and **methods (functions)**.



## Defining a Class

- Classes are defined using the `class` keyword.
- Special method `__init__()` is used for initialization.


##  __init__ Method

- `__init__` is a **constructor**.
- It runs automatically when an object is created.
- Used to initialize object attributes.

⚠️ Bad Practice:
Creating a class without `__init__` can make initialization unclear.


## self Keyword

- Represents the **current object**
- Used to access variables and methods inside the class

```python
# 1. Basic Class and Object

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")

s1 = Student("John", 21)
s1.display()
```

##  Methods in Class

- Methods are **functions defined inside a class**
- Used to perform operations on object data

## __str__ Method

- Returns a **user-friendly string representation**
- Used when printing an object


## __repr__ Method

- Returns a **developer-friendly representation**
- Used for debugging


```python
# 2. __str__ and __repr__

class Book:
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return f"Book: {self.title}"

    def __repr__(self):
        return f"Book('{self.title}')"

b = Book("Python Basics")
print(b)       # uses __str__
b              # uses __repr__ in notebook
```

##  __dict__

- Returns object attributes in **dictionary format**
- Useful for debugging and inspecting objects

```python
#  __dict__ Example

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Alice", 25)
print(p.__dict__)
```

##  Encapsulation

Encapsulation means **binding data and methods together** inside a class and restricting direct access.

Helps in:
- Data protection
- Controlled access


```python
# 3. Encapsulation Example

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # private variable

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

acc = BankAccount(1000)
acc.deposit(500)
print(acc.get_balance())
```

## Abstraction

Abstraction means **hiding unnecessary details** and showing only essential features.

Example:
You use a car without knowing how the engine works.



## ATM Machine Example

Abstraction hides the internal implementation and only shows the necessary features.

In an ATM:
- User interacts with buttons and screen
- User can withdraw money
- User can check balance

But the user does not know:
- How PIN verification happens
- How the bank database is accessed
- How transactions are processed

The complex implementation is hidden.

##  Inheritance

Inheritance allows a class (child) to **inherit properties and methods** from another class (parent).

 Benefits:
- Code reuse
- Reduces redundancy

##  Types of Inheritance

1. Single Inheritance - Single inheritance occurs when one child class inherits from one parent class.
2. Multiple Inheritance - Multiple inheritance occurs when one child class inherits from multiple parent classes.
3. Multilevel Inheritance - Multilevel inheritance occurs when a child class becomes a parent class for another class.
4. Hierarchical Inheritance - Hierarchical inheritance occurs when multiple child classes inherit from one parent class.
5. Hybrid Inheritance - Hybrid inheritance is a combination of two or more types of inheritance.

It can combine:

* Single inheritance
* Multiple inheritance
* Multilevel inheritance
* Hierarchical inheritance


```python
# 4. Inheritance Example

class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()
d.speak()
d.bark()
```

##  Method Overriding

- Child class provides a **different implementation** of a parent method
- Same method name, different behavior



```python
# 5. Method Overriding

class Animal:
    def speak(self):
        print("Animal sound")

class Dog(Animal):
    def speak(self):
        print("Bark")

d = Dog()
d.speak()
```

##  Polymorphism

Polymorphism means **same function name behaves differently** in different situations.

Example:
`len()` works for strings, lists, etc.


```python
# 6. Polymorphism Example

class Bird:
    def sound(self):
        print("Bird chirps")

class Dog:
    def sound(self):
        print("Dog barks")

def make_sound(animal):
    animal.sound()

make_sound(Bird())
make_sound(Dog())
```

## Arguments in Python

###  Positional Arguments (*args)
- Used to pass multiple values
- Stored as a tuple

###  Keyword Arguments (**kwargs)
- Used to pass named arguments
- Stored as a dictionary



```python
# 7. *args Example

def add_numbers(*args):
    return sum(args)

print(add_numbers(1, 2, 3, 4))
```

```python
# 8. **kwargs Example

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=25)
```
