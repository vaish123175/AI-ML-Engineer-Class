# Measures of Central Tendency

Measures of Central Tendency describe the **center** or **typical value** of a dataset. They help us understand where most of the data values are concentrated.

The three main measures of central tendency are:

- **Mean**
- **Median**
- **Mode**

## Mean (Average)

The **Mean** is the average of all the values in a dataset.

### Formula

$$
\text{Mean} = \frac{\text{Sum of all values}}{\text{Number of values}}
$$

### Example

Dataset:

```text
10 20 30 40 50
```

Calculation:

$$
\frac{10 + 20 + 30 + 40 + 50}{5} = \frac{150}{5} = 30
$$

**Mean = 30**

### When to Use

- Best used when there are **no extreme values (outliers)**.
- Commonly used for numerical data.


## Median

The **Median** is the middle value in a dataset after arranging the values in ascending or descending order.

### Example 1 (Odd Number of Values)

Dataset:

```text
5 10 15 20 25
```

The middle value is:

**Median = 15**

### Example 2 (Even Number of Values)

Dataset:

```text
10 20 30 40
```

There are two middle values (20 and 30).

$$
\text{Median} = \frac{20 + 30}{2} = 25
$$

**Median = 25**

### Median with Outliers

Dataset:

```text
10 15 20 25 500
```

- **Mean = 114**
- **Median = 20**

Although the value **500** is an outlier, the median remains **20**.

### When to Use

- Preferred when the dataset contains **outliers**.
- Commonly used for income, salaries, house prices, and other skewed data.


## Mode

The **Mode** is the value that appears **most frequently** in a dataset.

### Example

Dataset:

```text
2 3 3 5 6
```

Since **3** appears more often than the other values:

**Mode = 3**

### When to Use

- Useful for **categorical (qualitative)** data.
- Helps identify the most common value or category.

Examples:
- Most common blood group
- Most purchased product
- Most preferred brand




````python
print("Mean:",df["Marks"].mean())

print("Median:",df["Marks"].median())

print("Mode:")
print(df["Marks"].mode())
````

# Measures of Dispersion

Measures how spread out the data is.

## Types:
### Range 
Difference between the maximum and minimum values.
### Variance 
Variance measures the average squared distance from the mean.
- Small variance → values are close together.
- Large variance → values are spread out.
Because differences are squared, variance is expressed in squared units.
### Standard Deviation
Standard deviation is the square root of variance.
- It measures how far data points are from the mean in the original units.
- Low standard deviation → values are close to the mean.
- High standard deviation → values are widely spread.


````python
print("Maximum:",df["Marks"].max())

print("Minimum:",df["Marks"].min())

print("Range:",df["Marks"].max()-df["Marks"].min())

print("Variance:",df["Marks"].var())

print("Standard Deviation:",df["Marks"].std())
````

# Distribution

Distribution describes how the data is spread.

## Types:
### Symmetric Distribution
The left and right sides are mirror images.
Characteristics:
- Mean = Median = Mode
- Skewness = 0
The Normal Distribution is symmetrical.

### Asymmetric Distribution
One side has a longer tail than the other.
There are two types:
- Positive (Right) Skew
- Tail extends to the right.
- Mean > Median
- Example:
- Income data where a few people earn very high salaries.
- Negative (Left) Skew
- Tail extends to the left.
- Mean < Median
- Example:
A very easy exam where most students score highly.

A Normal Distribution is symmetric.

````python
x = np.random.normal(50,10,1000)

plt.hist(x,bins=30)

plt.title("Normal Distribution")

plt.xlabel("Values")

plt.ylabel("Frequency")

plt.show()
````

# Quartiles

Quartiles divide the dataset into four equal parts.

Q1 = 25%

Q2 = Median

Q3 = 75%

Inter Quartile Range (IQR) - Measures the spread of the middle 50% of the data.


IQR = Q3 - Q1

Used to detect outliers.

````python
Q1 = df["Marks"].quantile(0.25)

Q2 = df["Marks"].median()

Q3 = df["Marks"].quantile(0.75)

IQR = Q3-Q1

print("Q1 =",Q1)

print("Median =",Q2)

print("Q3 =",Q3)

print("IQR =",IQR)
````

# Outlier Detection

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Any value outside these bounds is considered an outlier.

````python
lower = Q1 - 1.5*IQR

upper = Q3 + 1.5*IQR

print("Lower Bound =",lower)

print("Upper Bound =",upper)

outliers = df[(df["Marks"]<lower) | (df["Marks"]>upper)]

print(outliers)
````

# Z-Score

Z-score tells how many standard deviations a value is from the mean.

Interpretation:
- Z = 0 → at the mean.
- Z = +2 → two standard deviations above the mean.
- Z = -1 → one standard deviation below the mean.

It is useful for identifying outliers and standardizing data.

````python
from scipy.stats import zscore

df["Z-Score"] = zscore(df["Marks"])

df
````

# Outlier Treatment
Outliers are unusually large or small values.
They can distort statistics and may cause machine learning models to overfit.
Common techniques include:
1. Capping (Winsorization)
Replace extreme values with a chosen lower or upper percentile (e.g., 1st and 99th percentiles) instead of removing them.
2. Binning
Group continuous values into ranges (bins).

Example:
- Age 0–18
- Age 19–35
- Age 36–60

This reduces the influence of extreme values.
Other techniques include removing outliers or transforming the data.

# Central Limit Theorem

The Central Limit Theorem states:
As the sample size becomes larger (typically 30 or more), the distribution of sample means approaches a normal distribution, regardless of the population's original distribution.

### Why it's important:
- Enables the use of many statistical tests.
- Makes sample statistics reliable estimates of population parameters.

````python
sample_means = []

for i in range(500):
    sample = np.random.randint(1,100,30)
    sample_means.append(np.mean(sample))

plt.hist(sample_means,bins=30)

plt.title("Central Limit Theorem")

plt.show()
````

# Hypothesis Testing

Hypothesis testing is a statistical method used to decide whether there is enough evidence to support a claim about a population.

Steps:
1. State the null hypothesis (H₀).
2. State the alternative hypothesis (H₁).
3. Choose a significance level (α), commonly 0.05.
4. Select an appropriate statistical test.
5. Calculate the test statistic and p-value.
6. Reject or fail to reject the null hypothesis.

## Types of Hypothesis Tests
### Z-Test
Used when:
- Sample size is large (typically ≥ 30).
- Population standard deviation is known.

Example:
Testing whether the average height of students differs from 170 cm.

### T-Test
Used when:
- Sample size is small (typically < 30).
- Population standard deviation is unknown.
Common types:
- One-sample t-test
- Independent two-sample t-test
- Paired t-test

### Chi-Square (χ²) Test
Used for categorical data.
Applications:
- Test whether two categorical variables are associated (independence test).
- Test whether observed frequencies match expected frequencies (goodness-of-fit test).

Example:
Checking whether gender and product preference are related.


````python
from scipy.stats import ttest_1samp

sample = [78,80,82,76,79,81,83]

result = ttest_1samp(sample,popmean=75)

print(result)
````

````python
from scipy.stats import chisquare

observed = [20,25,30,25]

expected = [25,25,25,25]

result = chisquare(observed,expected)

print(result)
````
