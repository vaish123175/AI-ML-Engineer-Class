## July 24: Pandas, NumPy & Matplotlib Notes

### Pandas

* Pandas is a Python library used for data analysis and manipulation.
* It is especially useful for:
* Cleaning data
* Filtering data
* Aggregating data
* Handling missing values
### Key Features
* Works with tabular (structured) data
* Supports CSV, Excel, SQL, JSON
* Provides powerful indexing and selection tools

### Data Structures in Pandas
### Series
* A one-dimensional labeled array
* Can store:
* Integers
* Strings
* Floats
* Example: A single column of data
### DataFrame
* A two-dimensional table
* Consists of:
* Rows (records)
* Columns (features)
* Each column can have different data types

```python
import pandas as pd

data = {
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, 30, 22],
    "Salary": [50000, 60000, 45000]
}

df = pd.DataFrame(data)
df
```

### Basic Functions
### .head()
* Returns first 5 rows (default)
* Helps in quick inspection


```python
df.head() 
df.head(2)
```

### .tail()
* Returns last 5 rows
* Useful for checking end of dataset

```python
df.tail() 
df.tail(2)
```

### .info()
* Provides:
* Column names

* Helps detect missing values

```python
print(df.info()) 
```

### .dtypes()
* Shows datatype of each column

```python
print(df.dtypes) 
```

### .describe()
* Statistical summary (numeric columns)
* Statistical summary of numeric columns:

* Mean → average
* Std → spread of data
* Min/Max → range
* Quartiles → distribution

```python
print(df.describe())
```

## Indexing in Pandas
### .iloc[] (Integer Location)
* Select data using index positions
* Syntax
* df.iloc[row_index, column_index]
* df.iloc[0]-> first row
* df.iloc[:, 1] -> second columnn

### .loc[] 
* Label-based indexing
* Uses column names instead of numbers

```python
## 1. Single value
df.iloc[0, 1]
```

```python
## 2. Entire row
df.iloc[1]
```

```python
## 3. Multiple rows
df.iloc[0:2]
```

```python
## 4. Specific column
df.iloc[:, 1]
```

```python
## 5. Rows + Columns
df.iloc[0:2, 0:2]
```

### Plotting (Matplotlib)
* Used for visualization (graphs)
* Library used for visualizing data
### Why Visualization?
* Helps understand:
* Trends
* Patterns
* Outliers

```python
import matplotlib.pyplot as plt

df["Salary"].plot(kind="bar")
plt.show()
```

### Types of Plots
* Line Plot
* Shows trends over time
* Bar Chart
* Compares categories
* Histogram
* Shows data distribution
* Pie Chart
* Shows proportions

### When to Use What
