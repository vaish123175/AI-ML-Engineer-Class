#  Encapsulation in OOP

## What is Encapsulation?

Encapsulation is an OOP concept that means **wrapping data (attributes) and methods (functions) together inside a single unit called a class**.

It also controls how the data can be accessed or modified.

Example:
A bank account contains:
- Account balance (data)
- Deposit and withdraw operations (methods)

The user cannot directly modify the balance. They must use methods.


##  Purpose of Encapsulation

Encapsulation helps in:

- Data protection
- Controlling access to variables
- Preventing accidental changes
- Improving code organization
- Making code easier to maintain


# Access Modifiers in Encapsulation

## Public Variables

Public variables can be accessed from anywhere.

Example:

self.name


## Protected Variables

Protected variables are represented using a single underscore (_).

They are intended to be accessed inside the class and subclasses.

Example:

self._salary

## Private Variables

Private variables are represented using double underscore (__).

They cannot be accessed directly outside the class.

Example:

self.__balance


Python uses name mangling to protect private variables.


```python
# Encapsulation Example

# Encapsulation Example

class BankAccount:
    
    def __init__(self, balance):
        # Private variable (cannot be accessed directly outside class)
        self.__balance = balance  

    def deposit(self, amount):
        # Adding amount to private balance
        self.__balance += amount  

    def get_balance(self):
        # Returning the private balance safely
        return self.__balance  


# Creating object with initial balance
account = BankAccount(1000)

# Calling method to deposit money
account.deposit(500)

# Accessing balance using method (not directly)
print(account.get_balance())
```

# Composition

## Definition

Composition is an OOP concept where one class contains an object of another class.

It represents a:

"Has-a relationship"


Example:

A Car has an Engine.

Car is not an Engine, but it contains an Engine object.

```python
# Composition Example

class Engine:
    
    def start(self):
        # Method inside Engine class
        print("Engine starts")


class Car:
    
    def __init__(self):
        # Car HAS an Engine (composition)
        self.engine = Engine()  

    def drive(self):
        # Using Engine's method through object
        self.engine.start()  
        print("Car is moving")


# Creating Car object
car = Car()

# Calling method
car.drive()
```

## Composition vs Inheritance


### Inheritance

Represents:

"IS-A relationship"


Example:

Dog is an Animal.


### Composition

Represents:

"HAS-A relationship"

Example:

Car has an Engine.

# Duck Typing

## Definition

Duck typing is a Python feature where the object's behavior matters more than its actual class.


Python checks:

"Does this object have the required method?"

instead of:

"What type of object is this?"


Example:

If an object has a fly() method,
Python treats it as something that can fly.


```python
# Polymorphism + Duck Typing Example

class Dog:
    
    def sound(self):
        # Dog-specific behavior
        print("Bark")


class Cat:
    
    def sound(self):
        # Cat-specific behavior
        print("Meow")


def make_sound(animal):
    # Function does not check type
    # It just calls sound() → duck typing
    animal.sound()


# Passing different objects
make_sound(Dog())   # Dog object
make_sound(Cat())   # Cat object
```

# Class Variables

## Definition

Class variables are variables declared inside the class but outside methods.


Characteristics:

- Shared by all objects
- Stored at class level
- Same value for every object
- Can be accessed using class name


Example:

ClassName.variable



# Instance Variables

## Definition

Instance variables belong to individual objects.


Characteristics:

- Created inside __init__()
- Different for every object
- Stored separately for each instance


Example:

self.name


```python
# Class Variable vs Instance Variable

class Student:
    
    school = "ABC School"   # Class variable (shared by all objects)

    def __init__(self, name):
        # Instance variable (unique for each object)
        self.name = name    


# Creating two objects
s1 = Student("John")
s2 = Student("Alice")

# Accessing instance variables
print(s1.name)   # John
print(s2.name)   # Alice

# Accessing class variable
print(s1.school) # Same for all
print(s2.school)
```

# Difference Between Class Variable and Instance Variable


| Class Variable | Instance Variable |
|---|---|
| Belongs to class | Belongs to object |
| Shared by all objects | Unique for each object |
| Created outside methods | Created inside __init__ |
| Uses class memory | Uses object memory |



# Methods in Classes


## Instance Method

Definition:

A method that works with object-specific data.

It uses:

self

as the first parameter.


Example:

student.display()


```python
# Instance Method

class Student:
    
    def __init__(self, name):
        # Assigning value to instance variable
        self.name = name  

    def display(self):
        # Instance method uses self to access data
        print(self.name)


# Creating object
student = Student("John")

# Calling instance method
student.display()
```

## Class Method

Definition:

A method that works with class-level data.

It uses:

cls

as the first parameter.


Created using:

@classmethod


```python
# Class Method

class Student:
    
    school = "ABC School"  # Class variable

    @classmethod
    def show_school(cls):
        # cls refers to class, not object
        print(cls.school)


# Calling class method using class name
Student.show_school()
```

## Static Method

Definition:

A method that does not access class or instance variables.

It behaves like a normal function inside a class.


Created using:

@staticmethod

```python
# Static Method

class Calculator:
    
    @staticmethod
    def add(a, b):
        # Static method does not use self or cls
        return a + b


# Calling static method
print(Calculator.add(10, 20))
```

# Decorators

## Definition

A decorator is a function that modifies another function's behavior without changing the original function code.


Advantages:

- Code reuse
- Cleaner code
- Adds extra functionality


Common uses:

- Logging
- Authentication
- Validation
- Measuring execution time



```python
# Decorator Example

def decorator(function):
    
    def wrapper():
        # Code before function call
        print("Before function")

        # Calling original function
        function()

        # Code after function call
        print("After function")

    return wrapper  # Returning modified function


@decorator
def hello():
    # Original function
    print("Hello World")


# Calling function (actually calls wrapper)
hello()
```

# Dataclass

## Definition

Dataclass is a Python feature that automatically generates common methods for classes.


It automatically creates:

- __init__()
- __repr__()
- __eq__()


Advantages:

- Less code
- Cleaner classes
- Easier object creation

```python
# Dataclass Example

from dataclasses import dataclass  # Importing dataclass

@dataclass
class Student:
    
    name: str   # Type hint
    age: int    # Type hint


# Automatically creates __init__ method
student = Student("John", 20)

# Automatically creates __repr__ method
print(student)
```

# Single Responsibility Principle (SRP)

## Definition

SRP states:

"A class should have only one reason to change."


Meaning:

A class should perform only one responsibility.


Example:

Bad Design:

User class:
- Store user data
- Send emails
- Generate reports


Better Design:

User class:
- Manage user information

Email class:
- Send emails


Report class:
- Generate reports


Benefits:

- Easier maintenance
- Better testing
- Cleaner code
- Less dependency


```python
# SRP (Single Responsibility Principle) Example

# Bad Design
class User:
    
    def save_user(self):
        # Saving user data
        print("Saving user")

    def send_email(self):
        # Sending email (different responsibility)
        print("Sending Email")


# ✅ Better Design

class User:
    
    def save(self):
        # Only responsible for saving user
        print("Saving user")


class Email:
    
    def send(self):
        # Only responsible for sending email
        print("Sending email")
```
