## Machine Learning

Machine Learning is a branch of Artificial Intelligence (AI) that allows computers to learn from data and make predictions or decisions without being explicitly programmed.

It is a technique where a system learns patterns from data and improves its performance automatically over time.

## How does it do?

1. Data Collection
Gather data from different sources
Data can be:
Numbers (sales, prices)
Text (emails, reviews)
Images (photos, videos)

2. Data Preprocessing
Clean and prepare data before training
Tasks include:

- Removing missing values
- Converting text to numbers
- Normalizing data

3. Choose a Model
Select an algorithm to learn patterns

Examples:

- Linear Regression
- Decision Tree
- Neural Network

4. Training the Model
- Feed data into the model
- Model learns relationships between input and output

model.fit(X_train, y_train)

5. Testing / Evaluation
- Check how well the model performs on new data

model.predict(X_test)

Metrics used:
- Accuracy
- Precision
- Recall

6. Prediction
Use trained model to predict new, unseen data

Example:

- Predict house price
- Classify email as spam or not

Data → Training → Model → Testing → Prediction

## Types of Machine Learning
 1. Supervised Learning
- Uses labeled data (input + correct output)
- Model learns mapping between input and output

Examples:

- House price prediction
- Spam email classification

2. Unsupervised Learning
- Uses unlabeled data
- Finds hidden patterns or groups

Examples:

- Customer segmentation
- Clustering

3. Reinforcement Learning
- Learns through trial and error
- Receives rewards or penalties

Examples:

- Game playing (chess, AI agents)
- Robotics

## 1. What is Linear Regression?

Linear Regression is a **supervised machine learning algorithm** used to model the relationship between:
- **Independent variable (X)** → input  
- **Dependent variable (Y)** → output  

It tries to find a **best-fit straight line** that predicts Y from X.

### Goal
To find a line that minimizes the difference between **actual values and predicted values**.


### Example
Predicting house price based on size:
- X → Size of house  
- Y → Price  

## 2. Line Equation

### Formula

y = mx + c

- **y** → Predicted output  
- **x** → Input feature  
- **m** → Slope (how much y changes with x)  
- **c** → Intercept (value of y when x = 0)  


###  Interpretation
- If m is positive → y increases as x increases  
- If m is negative → y decreases as x increases  





##  3. Uses of Regression

### Why use regression?

## 4. Gradient Descent

Gradient Descent is an **optimization algorithm** used to find the best values of m and c by minimizing error.

### Idea
- Start with random values of m and c  
- Calculate error  
- Update values to reduce error  
- Repeat until error is minimized  

## 5. R² (R-squared)

R² measures how well the model explains the variability of the data.

###  Formula

R² = 1 - (SS_res / SS_total)

### Interpretation
- **R² = 1** → Perfect fit  
- **R² = 0** → Model explains nothing  
- **Closer to 1 = better model**

## 6. Adjusted R²

Adjusted R² improves R² by adjusting for the number of features in the model.

### Formula
Adjusted R² = 1 - [(1 - R²)(n - 1) / (n - k - 1)]

###  Terms
- **n** → number of data points  
- **k** → number of features  



### Why Adjusted R²?
- R² always increases when adding features  
- Adjusted R² increases **only if feature improves the model**



## Key Differences

| Metric | Behavior |
|-------|--------|
| R² | Always increases with more features |
| Adjusted R² | Increases only if model improves
