# Transformers 

## 1. What Is a Transformer?

A **Transformer** is a neural network architecture designed to process
sequences of data, especially text. It became important in Natural
Language Processing (NLP) because it can learn relationships between
different parts of a sequence using a mechanism called
**self-attention**.

The Transformer architecture was introduced in the 2017 paper
**"Attention Is All You Need"**. The original Transformer was designed
as an **encoder-decoder architecture** for sequence-to-sequence tasks
such as machine translation.

The main idea behind a Transformer is:

> Instead of processing a sentence strictly one word at a time, the
> model can look at relationships between tokens using attention.

For example:

``` text
"The tiger jumped out of a tree because it was thirsty."
```

When the model processes the word **"it"**, it needs to understand that
"it" is related to **"the tiger"**. Self-attention allows the model to
calculate relationships between tokens and create a context-aware
representation.

# 2. Why Were Transformers Created?

Before Transformers, **Recurrent Neural Networks (RNNs)**, including
**LSTMs** and **GRUs**, were commonly used for sequence-processing
tasks.

An RNN processes a sequence sequentially:

``` text
I → love → machine → learning
```

The model processes one step and then uses information from that step
when processing the next step.

This creates difficulties for long sequences.

### Problems with RNNs

### 1. Sequential Processing

RNNs process tokens sequentially, which makes training harder to
parallelize.

``` text
Token 1
   ↓
Token 2
   ↓
Token 3
   ↓
Token 4
```

The next step depends on the previous step.

###  Long-Term Dependencies

When a sentence is very long, information from an earlier word can
become difficult to preserve.

For example:

``` text
"The student who studied very hard for several weeks
and completed all the assignments finally passed because
she understood the material."
```

The model needs to understand the relationship between:

``` text
student ↔ she
```

even though many words appear between them.

### Transformers

Transformers use self-attention to allow tokens to directly interact
with other tokens.

``` text
Token 1 ─┐
Token 2 ─┤
Token 3 ─┼──→ Self-Attention
Token 4 ─┘
```

This allows Transformers to model relationships over long sequences more
effectively and makes training more parallelizable than traditional RNN
processing.

## The Basic transformer process

Suppose I give the computer this sentence:

```text
The student studied hard because she wanted to pass.
```

A human immediately understands that **she** refers to **the student**. A computer cannot simply understand the sentence as English. It first has to convert the text into something a neural network can process.

So the Transformer follows:

```text
Sentence
   ↓
Break it into tokens
   ↓
Give every token an ID
   ↓
Turn the IDs into numerical embeddings
   ↓
Add information about position
   ↓
Use self-attention to understand relationships
   ↓
Process the information through Transformer layers
   ↓
Predict an output token
   ↓
Repeat until the response is complete
```

The most important idea in this entire structure is **attention**.

A Transformer basically asks:

> **"When I am trying to understand this token, which other tokens should I pay attention to?"**

For example, when it processes **she**, it can learn that **student** is important context. It does this using **self-attention**.

Think of a transformer as a pipeline :

``` text
   Text
    ↓
Tokenization
    ↓
Token Embeddings
    ↓
Positional Information
    ↓
Transformer Blocks
    ↓
Self-Attention
    ↓
   MLP
    ↓
Final Representation
    ↓
Linear Layer
    ↓
  Logits
    ↓
 Softmax
    ↓
Probabilities
    ↓
 Next Token
```

# What Is a Token?


The Transformer cannot directly send the English sentence into its neural-network calculations. So the first thing that happens is **tokenization**.


A token can be:

-   A complete word
-   Part of a word
-   Punctuation
-   A subword
-   Another piece of text

For example:

``` text
"I love machine learning."
```

could be represented as:

``` text
"I"
"love"
"machine"
"learning"
"."
```



# Token IDs

After tokenization, each token is mapped to an integer from the model's
vocabulary.

For example:

``` text
"I"        → 25
"love"     → 1934
"machine"  → 4521
"learning" → 6842
```

These are called **token IDs**.

The token ID itself does not contain the meaning of the word.

For example:

``` text
"cat" → 125
"dog" → 987
```

These tokens are then converted into numerical IDs. The IDs are simply labels that allow the model to identify each token.

But an ID such as `125` does not itself contain the meaning of a word. Therefore, the model looks up a learned **embedding vector** for each token.

# What Is an Embedding?

Neural networks work with numerical vectors rather than words.

Therefore, token IDs are converted into **embedding vectors**.

For example:

``` text
"cat"
   ↓
Token ID
   ↓
Embedding
   ↓
[0.21, 0.72, -0.14, 0.63, ...]
```

An embedding is a high-dimensional numerical representation of a token.

The embedding gives the model a much richer representation than the
token ID.

During training, the model learns useful representations for tokens.

# Token Embedding vs Token ID

These two concepts are different.

### Token ID

A token ID is an integer used to identify a token.

``` text
"cat" → 125
```

### Token Embedding

An embedding is a vector representing the token.

``` text
"cat" → [0.21, 0.72, -0.14, ...]
```

Therefore:

``` text
Token
  ↓
Token ID
  ↓
Embedding Vector
```

Now the computer has numerical representations that it can actually process.

There is still one problem: the model needs to know the order of the tokens. `Dog bites man` and `Man bites dog` contain the same words but have different meanings. Therefore, positional information is added.

So before attention begins,:

```text
Text
 ↓
Tokenization
 ↓
Token IDs
 ↓
Embeddings
 ↓
Positional Information
```


# Why Do We Need Positional Information?

Self-attention does not automatically give the model information about
the order of tokens.

Consider:

``` text
Dog bites man.
```

and:

``` text
Man bites dog.
```

The same words appear, but the meaning changes because the order
changes.

Therefore, the model needs information about the position of each token.

This is the purpose of **positional encoding / positional information**.

For example:

``` text
Token 1 → Position 1
Token 2 → Position 2
Token 3 → Position 3
```

The model can therefore consider both:

``` text
What is the token?
+
Where is the token?
```

So the model combines: 

``` text
Token Meaning
      +
Token Position
      ↓
Final Embedding
```

### Final Embedding

After token embedding and positional information are combined, we get the representation that is passed into the Transformer blocks.

Conceptually:
``` text 
Token
 ↓
Token Embedding
 +
Position Information
 ↓
Final Embedding
```

The final embedding contains information about both:

* What the token is
* Where the token appears

## What is a Transformer Block?

A Transformer model doesn't usually consist of just one layer.

It contains multiple Transformer blocks stacked on top of each other.

Each block processes and improves the token representations.

A simplified Transformer block contains:

``` text 
Input
 ↓
Self-Attention
 ↓
MLP
 ↓
Output
```

As information moves through the blocks, the representations become richer and more contextual.

# Input Preparation

A simplified input-preparation pipeline is:

``` text
Raw Text
   ↓
Tokenization
   ↓
Token IDs
   ↓
Token Embeddings
   ↓
Positional Information
   ↓
Transformer
```

According to the source material, input preparation can involve optional
text normalization, tokenization into words or subwords, conversion to
token IDs, conversion of token IDs into embeddings, and adding
positional information.


# What Is Self-Attention?

**Self-attention** is the central mechanism of a Transformer.

Self-attention allows each token to consider other tokens in the same
sequence and determine how relevant they are.

> **Self-attention is a mechanism in which each token can consider other
> tokens in the same sequence and determine how relevant they are. The
> model creates Query, Key, and Value representations, calculates
> Query-Key similarity scores, applies scaling and Softmax to obtain
> attention weights, and uses those weights to combine Value vectors.
> This creates context-aware representations of the tokens.**

Consider:

``` text
"The tiger jumped out of a tree to get a drink because it was thirsty."
```

When processing:

``` text
"it"
```

the model needs to understand its relationship with:

``` text
"the tiger"
```

Self-attention allows the model to calculate how strongly the tokens are
related.

The output is a **context-aware representation**.

# Context-Aware Representation

A token does not have to be represented in exactly the same way in every
context.

For example:

``` text
"I went to the bank to deposit money."
```

and:

``` text
"I sat beside the river bank."
```

The word:

``` text
bank
```

has different meanings depending on its surrounding context.

Self-attention helps the Transformer incorporate information from
surrounding tokens so the representation becomes more context-aware.

Therefore:

> **Self-attention helps the model understand a token based on its
> relationship with other tokens in the sequence.**

# Query, Key, and Value

Self-attention uses three important concepts:

``` text
Q = Query
K = Key
V = Value
```

A useful way to remember them is:

``` text
Query
→ What information am I looking for?

Key
→ What information can I be matched on?

Value
→ What information should I provide?
```

These are not literal questions being asked by the model. They are an
intuitive explanation of the mathematical operations.



## How Are Q, K, and V Created?

Each input embedding is transformed using learned weight matrices.

Conceptually:

``` text
Input Embedding
       ↓
   ┌───┼───┐
   ↓   ↓   ↓
  WQ  WK  WV
   ↓   ↓   ↓
   Q   K   V
```

The matrices are learned during model training.

The transformations create three different representations of the same
token:

``` text
Query
Key
Value
```

# Query

The **Query** represents what a token is looking for from the other
tokens.

For example, suppose the current token is:

``` text
"it"
```

The Query can be thought of as representing the type of information that
would help "it" determine which other token is relevant.

A useful mental model is:

> **Query = What am I looking for?**

# Key

The **Key** represents information used to determine whether another
token is relevant to a Query.

A useful mental model is:

> **Key = What kind of information do I represent that can be matched
> with a Query?**

The model compares a Query with Keys to calculate attention scores.


# Value

The **Value** contains the information that is actually passed forward
when a token receives attention.

A useful mental model is:

> **Value = What information should I receive from this token?**

Therefore:

``` text
Query + Key
      ↓
Determine relevance
      ↓
Value
      ↓
Retrieve information
```

The most important distinction is:

> **Query and Key determine relevance, while Value contains the
> information that gets combined.**


# Attention Score

The model compares a Query with Keys to determine how much attention
should be given to different tokens.

A common method is the **dot product**.

Conceptually:

``` text
Query × Key
     ↓
Attention Score
```

A higher score means the representations have a stronger match for the
attention calculation.

A lower score means a weaker match.

Therefore:

``` text
High Score
   ↓
More Attention
```

and:

``` text
Low Score
   ↓
Less Attention
```


# Scaled Dot-Product Attention

The standard attention operation is:

$$
Attention(Q,K,V)
=
Softmax
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
$$

The formula can be understood step by step.

``` text
QKᵀ
 ↓
Calculate similarity between queries and keys
 ↓
Divide by √dk
 ↓
Scale the scores
 ↓
Softmax
 ↓
Attention weights
 ↓
Multiply by V
 ↓
Attention output
```


### Why Do We Divide by √dk?

The attention scores are divided by:

``` text
√dk
```

where `dk` is the dimension of the key/query vectors.

This scaling helps keep the values at a suitable magnitude before
applying Softmax.

Without appropriate scaling, very large scores can make Softmax
extremely concentrated and make optimization more difficult.

For your exam, remember:

> **The division by √dk is used for numerical stability before applying
> Softmax.**


# Softmax in Attention

The attention scores are passed through the **Softmax function**.

Softmax converts the scores into attention weights.

For example:

``` text
Token A → 0.10
Token B → 0.60
Token C → 0.20
Token D → 0.10
```

The values sum to:

``` text
0.10 + 0.60 + 0.20 + 0.10 = 1.00
```
> **Softmax converts a set of scores into normalized
> probabilities/weights that sum to 1.**

Therefore, the model can interpret these as relative attention weights.

Token B receives the largest weight.

# Weighted Values

Once the attention weights are calculated, they are applied to the Value
vectors.

For example:

``` text
Token A → 0.10 × Value A
Token B → 0.60 × Value B
Token C → 0.20 × Value C
Token D → 0.10 × Value D
```

The weighted values are then added together.

This produces the new context-aware representation.

The complete process is:

``` text
Q + K
 ↓
Attention Scores
 ↓
Scale by √dk
 ↓
Softmax
 ↓
Attention Weights
 ↓
Weighted Values
 ↓
Attention Output
```

# Attention Matrix

When there are many tokens, the model does not calculate attention one
pair at a time manually.

The queries, keys, and values for all tokens can be organized into
matrices:

``` text
Q = Query Matrix
K = Key Matrix
V = Value Matrix
```

The attention operation can then be performed using matrix
multiplication.

This makes the operation efficient and suitable for parallel
computation.

# What Is Multi-Head Attention?

A Transformer does not rely on only one attention operation.

It uses **multiple attention heads**.

Each head has its own learned transformations for Query, Key, and Value.

Conceptually:

``` text
                    Input
                      ↓
        ┌─────────────┼─────────────┐
        ↓             ↓             ↓
      Head 1        Head 2        Head 3
        ↓             ↓             ↓
    Attention     Attention     Attention
        ↓             ↓             ↓
        └─────────────┼─────────────┘
                      ↓
                   Combine
                      ↓
                    Output
```

# Why Do We Need Multiple Attention Heads?

Different heads can learn different relationships within the input.

For example, one head might learn a relationship involving:

``` text
student ↔ she
```

Another might learn relationships involving:

``` text
verb ↔ subject
```

Another might capture a different relationship between words.

These examples are only intuitive. We should not assume that a specific
head always performs one easily interpretable task.

The important point is:

> **Multi-head attention allows the model to learn multiple types of
> relationships simultaneously.**

# How Multi-Head Attention Works

Each head performs its own attention calculation.

``` text
Head 1
Q1, K1, V1
   ↓
Attention 1
```

``` text
Head 2
Q2, K2, V2
   ↓
Attention 2
```

``` text
Head 3
Q3, K3, V3
   ↓
Attention 3
```

The outputs from the heads are then:

``` text
Concatenated
     ↓
Linear Transformation
     ↓
Multi-Head Attention Output
```

This gives the model a richer representation of the sequence.

# What Is the Feed-Forward Network?

After attention, the representation passes through a **Feed-Forward
Network**, also called an **MLP (Multi-Layer Perceptron)**.

The attention mechanism mainly allows tokens to exchange information.

The MLP then transforms each token's representation.

A simple way to remember it is:

``` text
Attention
→ Mix information between tokens

MLP
→ Process and transform the information
```

The feed-forward layer typically uses two linear transformations with a
nonlinear activation function such as **ReLU** or **GELU** between them.

# Residual Connections

Transformers contain many layers.

When a network becomes deep, it can become difficult to train because
information and gradients can become harder to propagate.

**Residual connections** provide a shortcut around a layer.

Conceptually:

``` text
Input ─────────────────────┐
  ↓                        │
Attention / MLP            │
  ↓                        │
Transformed Output ─── + ←─┘
```

The original input is added to the transformed output.

This is called a **residual connection** or **skip connection**.

# Why Are Residual Connections Important?

Residual connections help:

-   Information flow through deep networks
-   Gradient flow during training
-   Optimization
-   Reduce problems associated with vanishing and exploding gradients

A simple way to remember:

> **Residual connections provide shortcuts that make deep Transformer
> networks easier to train.**


# Layer Normalization

**Layer normalization** normalizes activations within a layer.

It helps stabilize the values flowing through the network.

Conceptually:

``` text
Input
 ↓
Transformation
 ↓
Normalization
 ↓
More Stable Representation
```

Layer normalization can improve gradient flow and help the model
converge more effectively during training.

For your notes:

> **Layer normalization helps stabilize the network and improve
> training.**

# Add & Norm

The original Transformer diagrams often show:

``` text
Add & Norm
```

This represents two operations:

``` text
Add
→ Residual / Skip Connection

Norm
→ Layer Normalization
```

So:

``` text
Input
 ↓
Attention
 ↓
Add & Norm
 ↓
Feed-Forward
 ↓
Add & Norm
```

This pattern is repeated throughout Transformer layers.

# What Is a Transformer Block?

A simplified Transformer block can be represented as:

``` text
Input
  ↓
Multi-Head Self-Attention
  ↓
Add & Norm
  ↓
Feed-Forward Network / MLP
  ↓
Add & Norm
  ↓
Output
```

Multiple Transformer blocks can be stacked together.

``` text
Input
 ↓
Block 1
 ↓
Block 2
 ↓
Block 3
 ↓
...
 ↓
Block N
 ↓
Final Representation
```

Each layer progressively transforms the representation.

# What Is an Encoder?

The original Transformer architecture contains an **encoder** and a
**decoder**.

The encoder receives the input sequence and creates contextualized
representations.

For example, in machine translation:

``` text
French Sentence
      ↓
Encoder
      ↓
Contextual Representation
```

The encoder's main purpose is:

> **Understand and represent the input sequence.**

The original Transformer uses multiple encoder layers.


# What Is a Decoder?

The decoder generates the output sequence.

For example:

``` text
French Sentence
      ↓
Encoder
      ↓
Representation
      ↓
Decoder
      ↓
English Translation
```

The decoder generates the output autoregressively, one token at a time.

Therefore:

``` text
Encoder
→ Understand / Represent Input

Decoder
→ Generate Output
```


# Original Transformer Architecture

The original Transformer was designed for sequence-to-sequence tasks
such as machine translation.

The architecture is:

``` text
Input Sequence
      ↓
Input Embedding
      ↓
Positional Encoding
      ↓
Encoder
      ↓
Contextual Representation
      ↓
Decoder
      ↓
Output Probability Distribution
      ↓
Output Sequence
```

For example:

``` text
French
"I love anime."
      ↓
Encoder
      ↓
Decoder
      ↓
English
"I love anime."
```

The actual source and target languages would differ in a translation
task.

# Encoder-Decoder Attention

In the original encoder-decoder Transformer, the decoder has access to
representations produced by the encoder.

This allows the decoder to use information from the input sequence while
generating the output.

For example:

``` text
Source Sentence
      ↓
Encoder
      ↓
Encoder Representation
      ↓
Decoder
      ↓
Target Sentence
```

This is especially useful for tasks where one sequence needs to be
transformed into another sequence.


#  GPT and the Original Transformer

It is important to distinguish between the **original Transformer** and
**GPT-style Transformers**.

### Original Transformer

``` text
Encoder + Decoder
```

The original Transformer was designed for sequence-to-sequence tasks
such as translation.

### GPT-style Transformer

``` text
Decoder-only
```

GPT-style models primarily use the decoder side of the Transformer
architecture and are designed for autoregressive language generation.

Therefore:

``` text
Original Transformer
→ Encoder + Decoder
```

``` text
GPT
→ Decoder-only Transformer
```

This is why diagrams of the original Transformer may show both encoder
and decoder, while GPT explanations focus heavily on masked
self-attention and next-token prediction.



# What Is Masked Self-Attention?

GPT-style models need to predict the next token.

Suppose the input is:

``` text
"I love"
```

The model needs to predict something such as:

``` text
"anime"
```

During training, the model should not be allowed to look at the future
token it is supposed to predict.

Therefore, future positions are **masked**.

Conceptually:

``` text
I
→ Can attend to I

love
→ Can attend to I and love

anime
→ When predicting the next token, future information must not be visible
```

The mask prevents information from future tokens from being used to
predict the current next token.



#  Why Is Masking Necessary?

Without masking, suppose the model is asked to predict:

``` text
"I love anime"
```

If it could see `anime` while trying to predict `anime`, it would
already know the answer.

That would make training invalid.

Masking forces the model to learn:

``` text
Previous Tokens
      ↓
Predict Next Token
```

instead of:

``` text
Previous + Future Tokens
      ↓
Predict Next Token
```


# What Is Autoregressive Generation?

**Autoregressive generation** means generating one token at a time,
where previously generated tokens become part of the context for
predicting the next token.

For example:

``` text
I love
   ↓
anime
```

Then:

``` text
I love anime
   ↓
because
```

Then:

``` text
I love anime because
   ↓
...
```

The process is:

``` text
Input
 ↓
Predict Next Token
 ↓
Add Predicted Token
 ↓
Predict Next Token
 ↓
Add Predicted Token
 ↓
Repeat
```

This is the basic generation process used by GPT-style language models.



# What Is a Language Model?

A language model predicts probabilities for possible next tokens based
on the previous context.

For example:

``` text
"The most famous city in the US is..."
```

The model might assign high probabilities to:

``` text
New York
Los Angeles
```

and low probabilities to unrelated tokens.

A modern Transformer-based language model learns these probability
distributions through training.

# What Is Next-Token Prediction?

A major training objective for decoder-only language models is
**next-token prediction**.

Suppose the training sentence is:

``` text
I love machine learning.
```

The training examples can conceptually be:

``` text
Input: I
Target: love

Input: I love
Target: machine

Input: I love machine
Target: learning

Input: I love machine learning
Target: .
```

The model learns to predict the next token from the previous context.

Over a huge amount of text, this task forces the model to learn many
patterns in language.

# Pre-Training

Large language models typically begin with a **pre-training** stage.

During pre-training, the model is exposed to a large amount of text,
often without manually provided labels for every example.

For an autoregressive language model, the task is to predict the next
token based on previous context.

Conceptually:

``` text
Large Text Dataset
       ↓
Tokenization
       ↓
Training Examples
       ↓
Predict Next Token
       ↓
Calculate Loss
       ↓
Update Model Parameters
       ↓
Repeat Many Times
```

The objective is to learn a general language representation and
probability distribution.


# What Are Model Parameters?

The model contains a very large number of learned numerical values
called **parameters**.

Examples include:

-   Embedding parameters
-   Attention weight matrices
-   Feed-forward network weights
-   Output-layer weights

During training:

``` text
Prediction
   ↓
Compare with Target
   ↓
Calculate Loss
   ↓
Backpropagation
   ↓
Update Parameters
```

The parameters gradually change so that the model becomes better at its
training objective.

# What Is a Logit?

After the Transformer processes the input, the model needs to produce
scores for possible next tokens.

These raw scores are called **logits**.

For example:

``` text
anime  → 4.2
music  → 2.1
pizza  → 1.5
school → 0.7
```

These values are not probabilities.

A larger logit means the model gives that token a higher score before
converting the scores into probabilities.


# Softmax for Output Prediction

The logits are passed through Softmax.

For example:

``` text
Logits:

anime  → 4.2
music  → 2.1
pizza  → 1.5
```

After Softmax:

``` text
anime  → 0.82
music  → 0.12
pizza  → 0.06
```

The probabilities sum to `1`.

The model now has a probability distribution over possible next tokens.


# Logits vs Probabilities

This distinction is important.

### Logits

``` text
anime → 4.2
music → 2.1
pizza → 1.5
```

Raw scores.

### Probabilities

``` text
anime → 0.82
music → 0.12
pizza → 0.06
```

Normalized values after Softmax.

Therefore:

``` text
Transformer
   ↓
Logits
   ↓
Softmax
   ↓
Probabilities
```


# How Does the Model Choose the Next Token?

After obtaining probabilities, the model needs a strategy for selecting
the next token.

Possible approaches include:

-   Greedy decoding
-   Sampling
-   Temperature
-   Top-K sampling
-   Top-P sampling

The choice affects how deterministic, diverse, or creative the output
can be.

# Greedy Decoding

Greedy decoding selects the token with the highest probability.

For example:

``` text
anime → 0.70
music → 0.20
pizza → 0.10
```

Greedy decoding chooses:

``` text
anime
```

The advantage is that it is simple and predictable.

However, always choosing the highest-probability token can sometimes
produce repetitive or less diverse text.

# Temperature

**Temperature** changes how concentrated or spread out the probability
distribution is during sampling.

### Low Temperature

``` text
Low Temperature
      ↓
More concentrated probabilities
      ↓
More predictable output
      ↓
Less randomness
```

### High Temperature

``` text
High Temperature
      ↓
More spread-out probabilities
      ↓
More varied output
      ↓
More randomness
```

Temperature does not retrain the model. It changes the way probabilities
are used during generation.


# Top-K Sampling

Top-K sampling restricts the model to the `K` most probable tokens.

Suppose:

``` text
dog     → 0.40
cat     → 0.30
rabbit  → 0.15
car     → 0.05
house   → 0.03
tree    → 0.02
```

If:

``` text
K = 3
```

only these candidates are considered:

``` text
dog
cat
rabbit
```

The model then samples from those candidates.

The important idea is:

> **Top-K keeps a fixed number of the most likely candidates.**


# Top-P Sampling

Top-P sampling uses cumulative probability instead of a fixed number of
tokens.

Suppose:

``` text
Top-P = 0.90
```

The model starts from the highest-probability tokens and keeps adding
candidates until the cumulative probability reaches approximately `90%`.

Therefore:

``` text
Top-K
→ Fixed number of candidates

Top-P
→ Variable number of candidates based on cumulative probability
```

Both methods are used to control the diversity of generated text.

## Step 1: Tokenization

The text is divided into tokens:

``` text
"I"
"love"
```

## Step 2: Token IDs

Each token is mapped to an ID:

``` text
"I" → Token ID
"love" → Token ID
```

## Step 3: Embeddings

The token IDs are converted into vectors:

``` text
"I" → [numbers...]
"love" → [numbers...]
```

## Step 4: Positional Information

The model adds position information:

``` text
"I" → Position 1
"love" → Position 2
```

## Step 5: Self-Attention

The model creates:

``` text
Q
K
V
```

and compares the Queries with Keys.

## Step 6: Attention Weights

Softmax converts attention scores into weights.

For example:

``` text
"I" → 0.30
"love" → 0.70
```

## Step 7: Weighted Values

The model combines the Value vectors according to these weights.

## Step 8: Multi-Head Attention

Several attention heads perform this operation in parallel.

## Step 9: MLP

The resulting representations are transformed through the feed-forward
network.

## Step 10: Add & Norm

Residual connections and layer normalization help stabilize the
processing and training.

## Step 11: Repeat

The representation passes through multiple Transformer blocks.

## Step 12: Logits

The final layer produces scores for possible next tokens:

``` text
anime → High Score
music → Medium Score
pizza → Low Score
```

## Step 13: Softmax

The logits are converted into probabilities:

``` text
anime → 0.70
music → 0.20
pizza → 0.10
```

## Step 14: Select Next Token

The model selects:

``` text
anime
```

Now the sequence becomes:

``` text
"I love anime"
```

The model can then use the expanded sequence to predict the next token.


# Why Are Transformers Powerful?

Transformers became powerful because they combine several useful ideas:

1.  **Self-attention**
    -   Learns relationships between tokens.
2.  **Multi-head attention**
    -   Learns multiple relationships simultaneously.
3.  **Positional information**
    -   Provides information about token order.
4.  **Feed-forward networks**
    -   Add nonlinear transformation and representation power.
5.  **Residual connections**
    -   Help information and gradients flow through deep networks.
6.  **Layer normalization**
    -   Helps stabilize training.
7.  **Stacked layers**
    -   Allow increasingly complex representations to be learned.

# Limitations of Transformers

Transformers are powerful, but they also have limitations.

One important limitation is the computational cost of self-attention.

For a sequence of length `n`, the standard self-attention mechanism
creates relationships between many pairs of tokens.

This leads to approximately quadratic computational and memory cost with
respect to sequence length:

``` text
O(n²)
```

Therefore, very long contexts can become expensive to process.

Modern Transformer systems use various techniques and architectural
improvements to address long-context efficiency.

# Transformer Architecture --- Big Picture

The entire concept can be remembered using this flow:

``` text
                 INPUT TEXT
                     ↓
                Tokenization
                     ↓
                  Token IDs
                     ↓
                 Embeddings
                     ↓
            Positional Information
                     ↓
              Transformer Blocks
                     ↓
        ┌────────────┴────────────┐
        ↓                         ↓
   Self-Attention               MLP
        ↓                         ↓
   Q, K, V                 Nonlinear Processing
        ↓                         ↓
   Attention Scores              │
        ↓                         │
     Scaling                     │
        ↓                         │
     Softmax                     │
        ↓                         │
 Attention Weights               │
        ↓                         │
 Weighted Values                 │
        └────────────┬────────────┘
                     ↓
                Add & Norm
                     ↓
              Repeat Layers
                     ↓
               Final Output
                     ↓
                  Logits
                     ↓
                 Softmax
                     ↓
             Probabilities
                     ↓
               Next Token
                     ↓
                  Repeat
```
