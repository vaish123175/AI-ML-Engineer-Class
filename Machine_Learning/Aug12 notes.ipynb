## 1. Drawbacks of Logistic Regression

Logistic Regression is mainly used for **classification problems**, especially when the relationship between the input features and the target can be separated reasonably well using a linear decision boundary.

### Main drawbacks

1. **Works mainly with linear decision boundaries**

   * Logistic Regression cannot naturally handle complex non-linear relationships.
   * Example: If Class A is inside a circle and Class B is outside the circle, a straight line cannot separate them properly.

2. **Sensitive to outliers**

   * Extreme values can affect the model's coefficients.

3. **Requires feature relationships to be reasonably suitable**

   * If the relationship between features and the target is highly complex, Logistic Regression may perform poorly.

4. **Feature engineering may be required**

   * We may need to create additional features to capture non-linear relationships.

5. **Multicollinearity can cause problems**

   * When input features are highly correlated with each other, the coefficients can become unstable.

### Example

Suppose we want to predict whether a customer will buy a product:

* Age
* Salary
* Previous purchases

If the relationship is approximately linear, Logistic Regression can work well.

But suppose the decision depends on a complicated combination such as:

> Young + low salary → Don't buy
> Middle age + medium salary → Buy
> Very high salary → Buy
> But certain combinations behave differently.

A **Decision Tree** may capture these rules more naturally.

# 2. Decision Tree

A **Decision Tree** is a supervised machine learning algorithm that can be used for:

* **Classification**
* **Regression**

It makes predictions by repeatedly asking questions about the data.

The structure looks like a tree:

```text
                 Income > $50K?
                  /          \
                Yes           No
                /              \
         Age > 30?            Don't Buy
          /     \
        Yes      No
        /         \
       Buy       Don't Buy
```

Each question creates a **split** in the data.


# 3. Why is it called a Decision Tree?

Because it has a tree-like structure.

### Important terms

**Root Node**

* The first/top node of the tree.
* It contains the complete dataset.

**Parent Node**

* A node that is split into smaller nodes.

**Child Node**

* The nodes created after splitting a parent node.

**Leaf Node**

* The final node where the prediction is made.

Example:

```text
             Age > 30?
             /       \
          Yes         No
          /            \
      Income > 50K     No
       /      \
     Yes       No
     /          \
    Buy       Don't Buy
```

Here:

* `Age > 30?` → Root node
* `Income > 50K?` → Child/parent node
* `Buy` → Leaf node
* `Don't Buy` → Leaf node



# 4. How Does a Decision Tree Work?

The Decision Tree makes decisions based on **conditions**.

For example:

```text
Is Age > 30?
```

If Yes:

```text
Is Salary > $50,000?
```

If Yes:

```text
Buy
```

If No:

```text
Don't Buy
```

The tree keeps splitting the data until it reaches a stopping condition.


# 5. Non-Linearity in Decision Trees

This is one of the biggest advantages of Decision Trees.

Decision Trees can learn **non-linear relationships**.

### Example

Imagine:

```text
         Class A
       A A A A A
     A A A A A A
     -------------
       B B B B
       B B B B
```

The model can create multiple rules to separate the classes.

A Decision Tree doesn't need one straight line to separate the data.

It can create rules such as:

```text
If Age < 25 → Class A

Else if Income > 50K → Class B

Else → Class A
```
Therefore, Decision Trees are good at handling complex decision patterns.


# 6. Pure Node and Impure Node

The goal of a Decision Tree is to create **pure nodes**.

## Pure Node

A node is pure when almost/all observations belong to the same class.

Example:

```text
Node:
Buy
Buy
Buy
Buy
Buy
```

This is a pure node because all observations belong to the same class.


## Impure Node

A node is impure when multiple classes are present.

Example:

```text
Buy
Don't Buy
Buy
Don't Buy
Buy
```

This node contains different classes, so it is impure.

### Goal

The Decision Tree tries to:

> **Reduce impurity after every split.**


# 7. Recursive Splitting

Decision Trees use **recursive splitting**.

This means:

1. Start with the complete dataset.
2. Find the best split.
3. Divide the dataset.
4. Find the best split for each resulting group.
5. Continue splitting.
6. Stop when a stopping condition is reached.

Example:

```text
                All Customers
                     |
              Age > 30?
              /       \
            Yes        No
            /           \
      Income > 50K       ...
        /     \
      Yes      No
```

The same process is repeatedly applied to the child nodes.


# 8. Greedy Approach

Decision Trees use a **greedy approach**.

At every step, the algorithm chooses the split that gives the **best immediate reduction in impurity**.

It does not look at every possible future tree and find the globally perfect tree.

It makes the best decision **at the current step**.

### Example

Suppose we have three possible features:

```text
Age       → impurity decreases by 20%
Income    → impurity decreases by 40%
Education → impurity decreases by 15%
```

The Decision Tree will choose:

```text
Income
```

because it provides the largest immediate reduction in impurity.


# 9. Problem with Too Much Splitting

If a Decision Tree keeps splitting until every small group becomes almost perfectly pure, it can **overfit**.

### Overfitting

The model learns the training data too closely, including noise and unusual observations.

Example:

```text
Training Accuracy = 99%
Testing Accuracy  = 70%
```

This is a sign that the tree may be overfitting.

### Why does this happen?

Because the tree becomes too complicated.

```text
Small dataset
     ↓
Many splits
     ↓
Very deep tree
     ↓
Learns noise
     ↓
Overfitting
```


# 10. High Variance

Decision Trees can have **high variance**.

High variance means the model can change significantly when the training data changes slightly.

For example:

```text
Training Dataset A
       ↓
Large Decision Tree

Training Dataset B
       ↓
Very different Decision Tree
```

This is one reason Decision Trees can overfit.


# 11. How Do We Handle Overfitting?

The main technique is **Pruning**.

Pruning means:

> Removing unnecessary branches from the Decision Tree.

There are two major approaches:

1. Pre-pruning
2. Post-pruning


# 12. Pre-Pruning

Pre-pruning means:

> Stop the tree from growing too much **during training**.

We define limits before the tree becomes too deep.

### Common hyperparameters

#### `max_depth`

Controls the maximum depth of the tree.

Example:

```text
max_depth = 3
```

The tree cannot grow beyond depth 3.


### `min_samples_split`

Specifies the minimum number of observations required to split a node.

Example:

```text
min_samples_split = 10
```

A node must have at least 10 observations before it can be split.


### `min_samples_leaf`

Specifies the minimum number of observations that must remain in a leaf.

Example:

```text
min_samples_leaf = 5
```

Every leaf must contain at least 5 observations.


### `max_leaf_nodes`

Limits the maximum number of leaf nodes.


## Why use pre-pruning?

Without restrictions:

```text
Tree keeps splitting
       ↓
Very deep tree
       ↓
Learns noise
       ↓
Overfitting
```

With pre-pruning:

```text
Set limits
    ↓
Tree stops earlier
    ↓
Simpler model
    ↓
Lower chance of overfitting
```



# 13. Post-Pruning

Post-pruning means:

> First allow the tree to grow, then remove unnecessary branches afterward.

Process:

```text
Grow large tree
      ↓
Evaluate branches
      ↓
Remove unnecessary branches
      ↓
Smaller tree
```

The idea is to keep branches that improve generalization and remove branches that mainly capture noise.


# 14. Goal of a Decision Tree

The main goal is:

> **Reduce impurity and create increasingly pure nodes.**

For classification, common impurity measures are:

* Entropy
* Gini Impurity

These are used to decide which split is better.


# 15. Entropy

**Entropy measures the amount of randomness or impurity in a dataset.**

### Low entropy

Data is mostly one class.

```text
Yes
Yes
Yes
Yes
```

Entropy is:

```text
0
```

This is completely pure.

### High entropy

Classes are mixed.

```text
Yes
No
Yes
No
```

There is more randomness.

For a binary classification problem, maximum entropy occurs when the classes are approximately equally distributed.


# 16. Entropy Formula

The formula is:

[
Entropy = -\sum p_i \log_2(p_i)
]

Where:

* (p_i) = probability of class (i)

### Simple example

Suppose we have 10 customers:

```text
6 Buy
4 Don't Buy
```

Therefore:

[
p(Buy)=0.6
]

[
p(Don't Buy)=0.4
]

So:

[
Entropy = -(0.6\log_2 0.6 + 0.4\log_2 0.4)
]

Approximately:

[
Entropy = 0.971
]

The closer the classes are to a 50/50 split, the higher the entropy.

# 17. Information Gain

**Information Gain measures how much entropy decreases after making a split.**

In simple words:

> Information Gain tells us how much randomness we removed by splitting the data.

Formula:

[
Information\ Gain =
Entropy(parent)
---------------

Weighted\ Entropy(children)
]

### Example

Suppose:

```text
Parent entropy = 1.0
```

After splitting:

```text
Weighted child entropy = 0.4
```

Then:

[
Information\ Gain = 1.0 - 0.4
]

[
Information\ Gain = 0.6
]

So the split reduced the uncertainty by **0.6**.

### Important

A good split generally has:

> **Higher Information Gain**

because it produces purer child nodes.

# 18. Gini Impurity

**Gini Impurity** is another way to measure impurity.

It is an alternative to entropy.

Formula:

[
Gini = 1-\sum p_i^2
]

For two classes:

[
Gini = 1-(p_1^2+p_2^2)
]

### Example

Suppose:

```text
6 Buy
4 Don't Buy
```

Then:

[
p_1=0.6
]

[
p_2=0.4
]

Therefore:

[
Gini = 1-(0.6^2+0.4^2)
]

[
=1-(0.36+0.16)
]

[
=0.48
]

So Gini impurity is:

[
\boxed{0.48}
]

### Interpretation

* **Gini = 0** → completely pure
* Higher Gini → more impurity

# 19. Entropy vs Gini

| Entropy                                 | Gini Impurity                   |
| --------------------------------------- | ------------------------------- |
| Measures randomness                     | Measures impurity               |
| Uses logarithm                          | Uses squared probabilities      |
| Used with Information Gain              | Used directly as Gini criterion |
| Slightly more computationally expensive | Usually faster                  |
| Lower entropy is better                 | Lower Gini is better            |

Both are commonly used to find good splits.


# 20. Decision Tree Algorithms

Different algorithms have been developed for building Decision Trees.

## ID3

**ID3 = Iterative Dichotomiser 3**

Main characteristics:

* Uses **Entropy**
* Uses **Information Gain**
* Originally designed mainly for classification
* Uses recursive splitting

Example:

```text
Choose feature with highest Information Gain
              ↓
Split data
              ↓
Repeat
```


# 21. C4.5

C4.5 is an improvement over ID3.

It uses:

> **Gain Ratio**

instead of simply Information Gain.

It can also handle:

* Continuous values
* Missing values
* Pruning

It is mainly used for classification.


# 22. CART

**CART = Classification and Regression Trees**

CART can be used for:

* Classification
* Regression

For classification, it commonly uses:

> **Gini Impurity**

For regression, it commonly uses a criterion based on reducing prediction error, such as squared-error reduction.

CART generally creates **binary splits**.

Example:

```text
Age > 30?
   /    \
 Yes     No
```


# 23. Decision Tree Example — Classification

# 24. Decision Tree for Regression

Decision Trees are not only for classification.

They can also predict numerical values.

Example:

Features:

* House size
* Number of bedrooms
* Location
* Age of house

The tree might look like:

# 25. Advantages of Decision Trees

### 1. Easy to understand

The decision rules are easy to interpret.

### 2. Handles non-linear relationships

It does not require a straight-line relationship.

### 3. Can be used for classification and regression

One algorithm family can solve both types of problems.

### 4. Little preprocessing

Decision Trees generally do not require feature scaling like:

* Standardization
* Normalization

### 5. Handles numerical and categorical features

Depending on the implementation.

### 6. Easy to visualize

The tree can be displayed and explained to non-technical users.


# 26. Disadvantages of Decision Trees

### 1. Overfitting

A deep tree can memorize the training data.

### 2. High variance

Small changes in the data can produce a very different tree.

### 3. Greedy approach

The locally best split isn't necessarily part of the globally best possible tree.

### 4. Can become very large

Without pruning, the tree can become complicated.

### 5. Sensitive to data changes

A small change in the dataset can change the tree structure.


# 27. Cross-Validation

Cross-validation is a technique used to evaluate how well a machine learning model will perform on **unseen data**.

Instead of depending on only one train/test split, we repeatedly train and evaluate the model on different portions of the data.

It helps us determine whether the model generalizes well.


# 28. Holdout Method

The simplest approach is the **Holdout Method**.

We divide the dataset into:

```text
Training Data
      +
Testing Data
```

Example:

```text
80% → Training
20% → Testing
```

The model learns from the 80%.

Then we evaluate it on the remaining 20%.

### Advantage

Simple and fast.

### Disadvantage

The result can depend heavily on how the data happened to be split.

# 29. K-Fold Cross-Validation

K-Fold Cross-Validation divides the dataset into **K groups/folds**.

For example:

```text
K = 5
```

Dataset:

```text
Fold 1
Fold 2
Fold 3
Fold 4
Fold 5
```

We train and test five times.

### Round 1

```text
Test → Fold 1
Train → Fold 2,3,4,5
```

### Round 2

```text
Test → Fold 2
Train → Fold 1,3,4,5
```

This continues until every fold has been used as the test set.

Finally:

> Calculate the average performance across all folds.

Example:

```text
Accuracy:
Fold 1 = 90%
Fold 2 = 88%
Fold 3 = 92%
Fold 4 = 89%
Fold 5 = 91%

Average = 90%
```

This gives a more reliable estimate than a single holdout split.


# 30. Leave-One-Out Cross-Validation

**LOOCV = Leave-One-Out Cross-Validation**

If we have 100 observations:

```text
99 → Training
1  → Testing
```

This is repeated 100 times.

Each observation gets used once as the test observation.

### Advantage

Uses almost all available data for training in each round.

### Disadvantage

Very computationally expensive for large datasets.


# 31. Stratified Split

Stratification is especially useful when the target classes are **imbalanced**.

Example:

Suppose we have:

```text
1000 customers

900 → No
100 → Yes
```

This is a class imbalance.

If we randomly split the data, we might accidentally get:

```text
Training:
950 No
50 Yes

Testing:
50 No
50 Yes
```

The testing data does not represent the original distribution.

### Stratified Split

Stratification maintains approximately the same class proportions in both sets.

Original:

```text
90% No
10% Yes
```

Training:

```text
90% No
10% Yes
```

Testing:

```text
90% No
10% Yes
```

This produces a more representative split.


# 32. Hyperparameter Tuning

A **hyperparameter** is a value we set **before training the model**.

It is not learned directly from the training data.

For Decision Trees, examples include:

```text
max_depth
min_samples_split
min_samples_leaf
max_leaf_nodes
criterion
```

The question becomes:

> Which combination of hyperparameters gives the best model?

This is called **Hyperparameter Tuning**.


# 33. Grid Search

**Grid Search** tries different combinations of hyperparameter values.

Example:

```text
max_depth:
3, 5, 7

min_samples_split:
2, 5, 10
```

Grid Search tests combinations such as:

```text
depth=3, split=2
depth=3, split=5
depth=3, split=10

depth=5, split=2
depth=5, split=5
depth=5, split=10

depth=7, split=2
depth=7, split=5
depth=7, split=10
```

It evaluates the combinations, often using **cross-validation**, and selects the combination with the best validation performance.


# 34. How These Concepts Connect

This is the important part to understand for interviews.

```text
Decision Tree
      ↓
Makes decisions using conditions
      ↓
Splits data recursively
      ↓
Uses impurity measure
      ↓
Entropy / Gini
      ↓
Finds the best split
      ↓
Greedy approach
      ↓
Tree keeps growing
      ↓
Can overfit
      ↓
Use pruning
      ↓
Pre-pruning / Post-pruning
      ↓
Control hyperparameters
      ↓
max_depth
min_samples_split
min_samples_leaf
      ↓
Evaluate model
      ↓
Cross-validation
      ↓
Holdout / K-Fold / LOOCV
      ↓
Tune hyperparameters
      ↓
Grid Search
```


# 35. Simple Real-World Example

Imagine a bank wants to predict whether a customer will get a loan.

Features:

```text
Age
Income
Credit Score
Existing Loans
```

The Decision Tree could create:

```text
                 Credit Score > 700?
                    /          \
                  Yes           No
                  /              \
          Income > $50K?       Reject
             /      \
           Yes       No
           /          \
        Approve      Reject
```

### What happened?

**Step 1:** The tree starts with all customers.

**Step 2:** It looks for the best feature to split the data.

**Step 3:** It may find that `Credit Score` gives the biggest impurity reduction.

**Step 4:** It splits the data.

**Step 5:** Each child node can be split again.

**Step 6:** It continues recursively.

**Step 7:** If it grows too much, it may overfit.

**Step 8:** We control the tree using pruning/hyperparameters.

**Step 9:** We use cross-validation to evaluate the model.

**Step 10:** We can use Grid Search to find good hyperparameter values.
