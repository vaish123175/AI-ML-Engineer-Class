## Machine Learning

Machine Learning is a branch of Artificial Intelligence (AI) that allows computers to learn from data and make predictions or decisions without being explicitly programmed.

It is a technique where a system learns patterns from data and improves its performance automatically over time.

## How does it do?

1. Data Collection
Gather data from different sources
Data can be:
Numbers (sales, prices)
Text (emails, reviews)
Images (photos, videos)

2. Data Preprocessing
Clean and prepare data before training
Tasks include:

- Removing missing values
- Converting text to numbers
- Normalizing data

3. Choose a Model
Select an algorithm to learn patterns

Examples:

- Linear Regression
- Decision Tree
- Neural Network

4. Training the Model
- Feed data into the model

5. Testing / Evaluation

- Accuracy

- Recall

Use trained model to predict new, unseen data

Example:

- Predict house price
- Classify email as spam or not

Data → Training → Model → Testing → Prediction

## 1. What is Linear Regression?

Linear Regression is a **supervised machine learning algorithm** used to model the relationship between:
- **Independent variable (X)** → input  
- **Dependent variable (Y)** → output  

It tries to find a **best-fit straight line** that predicts Y from X.

### Goal
To find a line that minimizes the difference between **actual values and predicted values**.

Examples:

* Predict house price
* Predict salary
* Predict sales
* Predict temperature

Linear regression assumes that there is a **linear relationship between the input features and the target/output**.

### Basic Formula

[
Y = mX + c
]

Where:

* `Y` = predicted output
* `X` = input feature
* `m` = slope/coefficient
* `c` = intercept

The model tries to find the **best-fit line** that represents the relationship between the input and output.

### Example

If:

* `X` = years of experience
* `Y` = salary

and salary generally increases as experience increases, linear regression can be used to model this relationship.

###  Interpretation
- If m is positive → y increases as x increases  
- If m is negative → y decreases as x increases  



##  2. Uses of Regression

- Easy to understand

# 3. Relationship Between X and Y

In simple linear regression:

* `X` = independent/input variable
* `Y` = dependent/target variable

There should generally be a **linear relationship** between X and Y.

This relationship can be checked using:

* Scatter plots
* Correlation
* Pearson correlation coefficient

A strong correlation does not automatically mean that linear regression will perform well, but it can indicate that a relationship exists.


# 4. Multiple Linear Regression

**Multiple Linear Regression** is used when there are **multiple input features** for predicting one target.

### Example

To predict house price, we might use:

* House size
* Number of bedrooms
* Number of bathrooms
* Location
* Age of house

The general formula is:

[
Y = b_0 + b_1X_1 + b_2X_2 + b_3X_3 + ... + b_nX_n
]

Where:

* `Y` = target
* `b₀` = intercept
* `b₁, b₂...` = coefficients
* `X₁, X₂...` = input features

##  Gradient Descent

Gradient Descent is an **optimization algorithm** used to find the best values of m and c by minimizing error.

### Idea
- Start with random values of m and c  
- Calculate error  
- Update values to reduce error  
- Repeat until error is minimized  

# 5. R² (R-Squared)

**R²** is a metric used to measure how well the regression model explains the variation in the target variable.

It is also called the **coefficient of determination**.

Generally:

* R² closer to `1` → model explains more variation
* R² closer to `0` → model explains less variation

For example:

If:

[
R^2 = 0.85
]

the model explains approximately **85% of the variation in the target**, according to the R² measure.

### Important

A higher R² does not always mean a better model.

Adding more features can increase R² even when those features are not useful.

# 6. Adjusted R²

**Adjusted R²** improves upon R² by taking the **number of input features** into account.

R² can increase when we add more variables, even if those variables do not meaningfully improve the model.

Adjusted R² penalizes unnecessary features.

### Why use Adjusted R²?

It is especially useful in **Multiple Linear Regression** when comparing models with different numbers of features.

###  Terms
- **n** → number of data points  
- **k** → number of features  



### Why Adjusted R²?
- R² always increases when adding features  
- Adjusted R² increases **only if feature improves the model**



### R² vs Adjusted R²

| R²                                        | Adjusted R²                                                      |
| ----------------------------------------- | ---------------------------------------------------------------- |
| Does not penalize unnecessary features    | Penalizes unnecessary features                                   |
| Usually increases when features are added | Can decrease when useless features are added                     |
| Good for measuring explained variation    | Better for comparing models with different numbers of predictors |



# 7. Multicollinearity

**Multicollinearity** occurs when two or more independent/input variables are highly correlated with each other.

### Example

Suppose we have:

* House size in square feet
* House size in square meters

These two features provide almost the same information.

They are highly correlated with each other.

This creates multicollinearity.



# 8. Why Should We Care About Multicollinearity?

Multicollinearity can cause problems in linear regression because the model has difficulty determining the individual effect of each correlated feature.

It can cause:

* Unstable coefficients
* Large standard errors
* Difficult interpretation of coefficients
* Coefficients changing significantly when data changes
* Less reliable conclusions about individual features

### Important

Multicollinearity mainly affects the **interpretation and stability of coefficients**. It does not necessarily mean that the model's predictions will always be poor.



# 9. Target Correlation vs Feature Correlation

There are two different relationships to examine.

### Feature → Target

We want useful input features to have some meaningful relationship with the target.

Example:

`House Size → House Price`

### Feature → Feature

We should check whether input features are highly correlated with each other.

Example:

`House Size in sq ft ↔ House Size in sq meters`

This second relationship is where **multicollinearity** becomes a concern.



# 10. Pearson Correlation

**Pearson correlation** measures the strength and direction of a **linear relationship** between two numerical variables.

The correlation coefficient ranges from:

[
-1 \text{ to } +1
]

### Interpretation

* `+1` → perfect positive linear relationship
* `0` → no linear relationship
* `-1` → perfect negative linear relationship

### Positive Correlation

When one variable increases, the other tends to increase.

### Negative Correlation

When one variable increases, the other tends to decrease.



# 11. Pearson Correlation Heatmap

A **correlation heatmap** is a visual representation of correlations between multiple numerical variables.

It helps us quickly identify:

* Features strongly correlated with the target
* Features strongly correlated with each other
* Potential multicollinearity

For example:

If two input features have a correlation close to `+1` or `-1`, they may have a multicollinearity problem.



# 12. Remedies for Multicollinearity

There are several ways to handle multicollinearity.

### 1. Do Nothing

If multicollinearity does not significantly affect the model or interpretation, we may leave it as it is.

### 2. Remove One Variable

If two features provide similar information, remove one of them.

Example:

* `House size in sq ft`
* `House size in sq meters`

Keep one and remove the other.

### 3. Combine Variables

Related variables can sometimes be combined into a new feature.

### 4. Principal Component Analysis (PCA)

**PCA** transforms correlated features into a smaller number of new, uncorrelated features called **principal components**.

PCA can help reduce the effect of multicollinearity while preserving important information from the original variables.

### 5. Regularization

Techniques such as **Ridge Regression** can help deal with multicollinearity by controlling the size of coefficients.



# 13. Bias-Variance Tradeoff

The **Bias-Variance Tradeoff** is an important concept in machine learning.

The goal is to build a model that has a good balance between:

* Bias
* Variance

We want a model that generalizes well to **unseen data**.



# 14. Bias

**Bias** is the error caused by a model being too simple and making strong assumptions about the data.

A high-bias model cannot capture the important patterns in the data.

### High Bias

Usually leads to:

**Underfitting**

Example:

Trying to use a straight line to represent a relationship that is strongly curved.



# 15. Variance

**Variance** refers to how sensitive a model is to changes in the training data.

A high-variance model learns the training data too closely.

### High Variance

Usually leads to:

**Overfitting**

The model performs very well on training data but poorly on new/unseen data.



# 16. Bias vs Variance

| Situation | Bias | Variance | Result       |
| --------- | ---: | -------: | ------------ |
| High Bias | High |      Low | Underfitting |
| Low Bias  |  Low |     High | Overfitting  |
| Low Bias  |  Low |      Low | Good model   |
| High Bias | High |     High | Poor model   |

### Important

The goal is **not simply to make both values as low as possible**. In practice, we try to find a model complexity that gives good generalization with an appropriate balance.



# 17. Total Prediction Error

A simplified decomposition of expected prediction error is:

[
Total\ Error = Bias^2 + Variance + Irreducible\ Error
]

### Bias²

Error caused by incorrect or overly simplistic assumptions.

### Variance

Error caused by the model being too sensitive to the training data.

### Irreducible Error

Noise that exists naturally in the data and cannot be completely removed by the model.


# 18. Overfitting

**Overfitting** occurs when a model learns the training data too closely, including noise and random patterns.

### Characteristics

* Very good training performance
* Poor test performance
* High variance
* Model is too complex

### Example

A model memorizes the training examples instead of learning general patterns.


# 19. Underfitting

**Underfitting** occurs when a model is too simple to learn the important patterns in the data.

### Characteristics

* Poor training performance
* Poor test performance
* High bias
* Model is too simple

### Example

Using a simple straight line when the actual relationship between variables is highly nonlinear.


# 20. Bias-Variance Tradeoff

As model complexity increases:

* Bias generally decreases
* Variance generally increases

As model complexity decreases:

* Bias generally increases
* Variance generally decreases

The goal is to find a suitable **tradeoff point** where the model generalizes well.



# 21. Regularization

**Regularization** is a technique used to reduce overfitting by adding a penalty for large model coefficients.

It encourages the model to prefer simpler solutions.

### Why use Regularization?

Regularization helps:

* Reduce overfitting
* Control model complexity
* Reduce very large coefficients
* Improve generalization
* Handle multicollinearity in some cases

Regularization is commonly used with:

* Ridge Regression
* Lasso Regression
* Elastic Net



# 22. Linear Regression Without Regularization

The basic linear regression model tries to minimize a loss such as **Mean Squared Error (MSE)**.

Conceptually:

[
Loss = MSE
]

With regularization, a penalty term is added:

[
Loss = MSE + Regularization\ Penalty
]

This encourages the model to keep coefficients under control.



# 23. Ridge Regression – L2 Regularization

**Ridge Regression** uses **L2 regularization**.

The penalty is based on the squared coefficients:

[
Loss = MSE + \lambda\sum_{j=1}^{p}\beta_j^2
]

Where:

* `MSE` = prediction error
* `λ` = regularization strength
* `β` = model coefficients

### What Ridge Does

Ridge **shrinks coefficients toward zero**, but generally does not make them exactly zero.

Therefore, Ridge usually keeps all features in the model.



# 24. Lasso Regression – L1 Regularization

**Lasso Regression** uses **L1 regularization**.

The penalty is based on the absolute value of coefficients:

[
Loss = MSE + \lambda\sum_{j=1}^{p}|\beta_j|
]

### What Lasso Does

Lasso can make some coefficients exactly:

[
0
]

When a coefficient becomes zero, that feature effectively does not contribute to the model.

Therefore, Lasso can perform **feature selection**.



# 25. Ridge vs Lasso

| Ridge                              | Lasso                                        |
| ---------------------------------- | -------------------------------------------- |
| L2 regularization                  | L1 regularization                            |
| Shrinks coefficients               | Can shrink coefficients to exactly zero      |
| Usually keeps all features         | Can remove features                          |
| Good when many features contribute | Useful when some features may be unnecessary |
| Helps with multicollinearity       | Can perform feature selection                |

### Easy way to remember

**Ridge = Shrink**

**Lasso = Remove**



# 26. Elastic Net

**Elastic Net** combines both:

* L1 regularization from Lasso
* L2 regularization from Ridge

Therefore:

[
Loss = MSE + L1\ Penalty + L2\ Penalty
]

Elastic Net can:

* Shrink coefficients
* Set some coefficients to zero
* Handle correlated features better than Lasso in many situations



# 27. When Should We Use Each Regression?

### Plain Linear Regression

Use when:

* Relationship is approximately linear
* Number of features is manageable
* Multicollinearity is not a major concern
* Overfitting is not a major issue

### Ridge Regression

Use when:

* There are many features
* Features are correlated
* You want to keep most/all features
* You want to reduce coefficient size

### Lasso Regression

Use when:

* You suspect many features are unnecessary
* You want automatic feature selection
* You want some coefficients to become zero

### Elastic Net

Use when:

* You have many features
* Features are correlated
* You want both feature selection and coefficient shrinkage
* Lasso alone may be unstable with groups of correlated features



# 28. Polynomial Regression

Real-world relationships are not always linear.

Sometimes the relationship between input and output is **curved**.

In such cases, Polynomial Regression can be used.

### Example

Instead of:

[
Y = b_0 + b_1X
]

we can use:

[
Y = b_0 + b_1X + b_2X^2 + b_3X^3 + ... + b_nX^n
]

The **degree** determines the highest power of X used.



# 29. Polynomial Degree

The degree controls the complexity of the model.

### Degree 1

[
Y = b_0 + b_1X
]

This is essentially ordinary linear regression.

### Degree 2

[
Y = b_0 + b_1X + b_2X^2
]

This can represent a curved relationship.

### Degree 3

[
Y = b_0 + b_1X + b_2X^2 + b_3X^3
]

This can represent a more complex curve.


# 30. Why Use Polynomial Regression?

We use Polynomial Regression when the relationship between the features and target is **nonlinear**, but we still want to model it using regression.

Examples:

* Growth patterns
* Temperature relationships
* Sales trends
* Physical measurements
* Certain economic relationships

The goal is to find a curve that best represents the underlying relationship.


# 31. Polynomial Regression and Overfitting

Increasing the polynomial degree makes the model more flexible.

However, a very high degree can cause **overfitting**.

### Low Degree

Model may be too simple.

→ **Underfitting**

### Very High Degree

Model may follow noise in the training data.

→ **Overfitting**

### Appropriate Degree

Captures the important pattern without following random noise.

→ **Better generalization**


# 32. How to Handle Underfitting

If the model is underfitting:

* Use a more appropriate model
* Increase model complexity when justified
* Add useful features
* Use polynomial features when the relationship is nonlinear
* Reduce excessive regularization
* Train with better-quality features



# 33. How to Handle Overfitting

If the model is overfitting:

* Use regularization
* Reduce model complexity
* Reduce polynomial degree
* Remove unnecessary features
* Collect more training data
* Use cross-validation
* Perform feature selection
* Use an appropriate model
