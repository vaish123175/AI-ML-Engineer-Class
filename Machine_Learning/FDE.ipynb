# Forward Deployed Engineering (FDE)

## What is Forward Deployed Engineering?

Forward Deployed Engineering is an engineering approach where engineers work closely with customers or organizations to take a technology, software product, or AI capability and make it work in the customer's real-world environment.

A normal software engineer may mainly work on the company's core product. A Forward Deployed Engineer works much closer to the customer and focuses on questions such as:

- What problem is the customer actually trying to solve?
- What data does the customer have?
- What systems are they already using?
- How can the new technology connect to those systems?
- How can the solution be deployed safely?
- How do we know whether the solution is successful?
- How can the solution be improved after deployment?

The important idea is that an FDE does not simply build something that works on a developer's laptop. The FDE helps move the solution from **"the technology works"** to **"the technology works for this particular customer in production."**

## Why Do We Need FDE?

Imagine that a company develops a powerful AI model. In a demonstration, everything works perfectly.

```text
AI Model
   ↓
Demo
   ↓
Works!
```

So the AI model that worked perfectly in a simple demo may not work immediately in the customer's real environment.

```text
AI Model
   ↓
Customer's real environment
   ↓
Different data
Different systems
Different security
Different workflows
   ↓
Engineering problem
```

For example, imagine the AI expects data in this format:

Name | Age | Department

But the customer's database stores the same information differently:

employee_name | dept_name | employee_age

The AI model itself may be working perfectly, but now someone has to figure out how to connect the AI to the customer's data.

This is where a Forward Deployed Engineer becomes important.

The FDE asks:

"How can I make this technology work in the customer's environment?"

The FDE may need to:

* Connect the AI to the customer's database
* Work with existing APIs
* Convert or clean the customer's data
* Handle security requirements
* Work with legacy systems
* Integrate the AI into existing applications
* Deploy the solution
* Test it with real customer data
* Fix problems that appear in the real environment

So the important difference is:

``` text
Technology works in a demo
             ↓
          Not enough
             ↓
Technology must work with
the customer's real environment
             ↓
             FDE
             ↓
       Working Solution
```

This is why we need Forward Deployed Engineers.

The FDE acts as a bridge between the technology company and the customer's real environment.
   

# Basic FDE Lifecycle


```text
Understand the Problem
        ↓
Understand the Customer
        ↓
Understand the Data
        ↓
Design the Solution
        ↓
Build a Prototype
        ↓
Integrate with Existing Systems
        ↓
Test and Evaluate
        ↓
Deploy to Production
        ↓
Monitor
        ↓
Improve
        ↓
Deliver Business Value
```

This represents the journey from a customer problem to a working production solution.


# Step 1 — Understand the Problem

The first and most important step for an FDE is to **understand the customer's problem before writing any code**.

An FDE should not immediately hear a problem and start programming. First, they need to understand **what the customer actually needs**.

For example, imagine a hospital says:

> "We want to use AI to help our doctors."

This sounds good, but it is too general. The FDE needs to ask more questions.

The FDE might ask:

- What is taking too much time?
- Are doctors spending too much time reading documents?
- Is it difficult to find patient information?
- Do doctors need help creating reports?
- Do they need an AI system to answer questions?
- What software and systems does the hospital already use?
- What would make the AI solution successful?

The FDE keeps asking questions until the problem becomes clear.

For example:

```text
Vague Problem:

"We want AI."
       ↓
FDE asks questions
       ↓
Find the real problem
       ↓
Specific Problem:

"Doctors spend too much time
searching through patient documents."
       ↓
Technical Problem:

"How can we help doctors
quickly find relevant information?"
       ↓
Possible Solution:

"Build a secure system that helps
doctors search relevant documents."
``` 

So, the FDE's job is to take a big and unclear business problem and turn it into a clear technical problem that engineers can solve.

This ability to understand the customer's needs and translate them into technical requirements is one of the most important skills of a Forward Deployed Engineer.

# Step 2 — Understand the Customer's Environment

After understanding the problem, the FDE needs to understand where the solution will work.

For example:

```text
Customer
   ↓
Database
   ↓
APIs
   ↓
Cloud
   ↓
Existing Applications
   ↓
Security Systems
   ↓
Users
```

The FDE needs to understand how these parts connect and work together.

This is why FDEs need broader knowledge than just one programming language.

They may need knowledge of:

- Python
- SQL
- APIs
- Cloud platforms
- Databases
- Docker
- Git
- CI/CD
- Authentication
- Networking
- Machine Learning
- LLMs
- RAG
- Monitoring
- System architecture

The FDE does not necessarily need to be an expert in everything, but they should understand how these technologies fit together to build a working solution.

# Step 3 — Understand the Data

For AI projects, data is extremely important.

A customer may have data in many different places:

```text
Customer Data

CSV files
   +
SQL databases
   +
PDF documents
   +
APIs
   +
Internal applications
```

However, the data may have problems:

- Missing data
- Duplicate data
- Inconsistent data
- Incorrect formatting
- Data stored in different systems
- Data that is difficult to access

So the FDE needs to ask:

**Can we actually use this data to solve the customer's problem?** 

This is called **data readiness**.

Before building an AI solution, the FDE needs to make sure the required data is available, accessible, and useful.

# Step 4 — Design the Solution

Once the FDE understands the **problem, environment, and data**, they can design the technical solution.

For example, suppose a company wants an AI assistant that can answer questions using its internal documents.

The FDE might design:

```text
Company Documents
       ↓
Data Processing
       ↓
    Chunking
       ↓
   Embeddings
       ↓
Vector Database
       ↓
   Retriever
       ↓
      LLM
       ↓
     Answer
```

This is a common **RAG (Retrieval-Augmented Generation)** architecture.

The FDE needs to decide:

- Which AI model should we use?
- Where should the data be stored?
- Which vector database should we use?
- How should documents be divided into chunks?
- How should the system retrieve information?
- How should users authenticate?
- How will the system connect to existing applications?
- How will we evaluate the AI responses?
- Where will the system be deployed?

The goal is to design a solution that is technically possible, useful for the customer, secure, and scalable.


# Step 5 — Build a Prototype

The FDE usually does not build the complete production system immediately.

Instead, they may create a **Proof of Concept (POC)** or prototype.

The purpose is to answer:

> **"Can this actually solve the customer's problem?"**

For example:

```text
Customer Problem
       ↓
Small Prototype
       ↓
Test with Real Data
       ↓
Evaluate Results
       ↓
Does it work?
```

If it does not work, the FDE changes the approach.

For example, they might:

- Try a different AI model
- Improve the data
- Change the retrieval method
- Change the architecture
- Improve the prompts
- Use a different database

This is important because the FDE does not want to spend months building a complete system only to discover later that it does not solve the customer's actual problem.


# Step 6 — Integration

**Integration** is one of the most important parts of an FDE's job.

After building and testing the solution, the FDE needs to **connect it to the customer's existing systems**.

The AI solution should not work separately. It needs to communicate with the systems the customer is already using.

For example:

```text
Customer Database
       ↓
      API
       ↓
AI Application
       ↓
   LLM / RAG
       ↓
Customer Interface
```

The FDE may need to build or connect:

* **APIs** — Allow different applications to communicate with each other.
* **Database connections** — Allow the AI application to read or write customer data.
* **Authentication** — Make sure only authorized users can access the system.
* **Data pipelines** — Move and prepare data between different systems.
* **Backend services** — Handle the application's business logic.
* **Frontend components** — Provide an interface for customers or employees to use the solution.
* **Cloud infrastructure** — Run the application and AI services in the cloud.
* **Automation** — Automatically perform repetitive tasks or workflows.

For example, imagine a company already has a customer database.

The FDE builds an AI assistant that needs information from that database.

The FDE cannot simply build the AI assistant and stop there.

They need to connect:

```text
Existing Customer Database
          ↓
       API / Backend
          ↓
      AI Application
          ↓
        LLM / RAG
          ↓
     Customer Interface
```

Now the AI application can work with the customer's existing technology.

This is why FDE work is often described as **integration engineering** as well as **software engineering**.

> **Build the solution → Connect it to existing systems → Make everything work together.**

The FDE's goal is not just to build an AI application.

The goal is to make the **entire solution work inside the customer's real environment**.

# Step 7 — Evaluation

Just because an AI system **produces an answer** does not mean that the answer is correct or useful.

The FDE needs to **test and evaluate the system** to make sure it is actually solving the customer's problem.

The basic process looks like this:

```text
Question
   ↓
AI System
   ↓
Answer
   ↓
Evaluation
```

During evaluation, the FDE asks important questions such as:

* **Is the answer correct?**
* **Did the AI retrieve the right information?**
* **Did the model hallucinate or make up information?**
* **Is the response fast enough?**
* **Is the system reliable?**
* **Is the customer's data secure?**
* **Does the system meet the customer's requirements?**

For example, imagine a customer asks:

> "What is our company's refund policy?"

The AI system retrieves information from company documents and generates an answer.

The FDE needs to check:

```text
Customer Question
       ↓
Did the system find the correct document?
       ↓
Did it retrieve the correct information?
       ↓
Did the LLM understand the information?
       ↓
Is the final answer correct?
       ↓
Does it answer the customer's question?
```

If the answer is wrong, the FDE needs to find **where the problem happened**.

Maybe:

* The wrong document was retrieved.
* The document was not properly chunked.
* The embedding was not effective.
* The LLM misunderstood the retrieved information.
* The prompt needs improvement.
* The data itself was incorrect.

This is why evaluation is especially important in **AI systems**.

A system can technically work:

```text
Question → AI → Answer
```

but still produce **incorrect, slow, unreliable, or unsafe answers**.

The FDE's goal is not simply to prove that the AI system works.

The goal is to prove that it works well enough to solve the customer's real problem.


# Step 8 — Production Deployment

**Production deployment** is the stage where the solution moves from:

> **"It works on my laptop."**

to:

> **"Real users can use it."**

During development, an application may work perfectly on the FDE's computer.

But a real customer needs the application to work for **many users, securely and reliably**.

The deployment process looks like this:

```text id="8y0j0c"
Prototype
   ↓
Testing
   ↓
Security Review
   ↓
Deployment
   ↓
Production
   ↓
Real Users
```

Before deploying the system, the FDE needs to think about several important areas:

* **Reliability** — Does the system continue working without frequent failures?
* **Scalability** — Can the system handle more users and more data?
* **Security** — Is customer information protected?
* **Authentication** — Can only authorized users access the system?
* **Monitoring** — Can we see what is happening inside the application?
* **Performance** — Is the system fast enough for users?
* **Cost** — How much will it cost to run the system?
* **Failure handling** — What happens if something goes wrong?
* **Deployment infrastructure** — Where and how will the application run?

For example, imagine an FDE built an AI assistant on their laptop.

During development:

```text id="s4nq4v"
FDE's Laptop
     ↓
AI Application
     ↓
    LLM
     ↓
   Answer
```

That is enough for testing, but it is **not enough for a real customer**.

For production, it might become:

```text id="w4z2ke"
Real Users
     ↓
Web Application
     ↓
Cloud Infrastructure
     ↓
Backend / API
     ↓
RAG System
     ↓
    LLM
     ↓
Database / Vector Database
```

The FDE also needs to monitor the system after deployment.
`
For example:

```text id="i2j7v8"
Production System
       ↓
   Monitoring
       ↓
Are there errors?
       ↓
Is the system slow?
       ↓
Are costs increasing?
       ↓
Are users getting good answers?
```

If problems occur, the FDE investigates and improves the system.

This is why an FDE benefits from knowledge of **DevOps and cloud infrastructure**, in addition to programming and AI.


# Step 9 — Monitoring

**Deployment is not the end.**

Once the system is running in production, the FDE needs to **monitor it continuously** to make sure everything is working properly.

```text id="q6s5pe"
Production AI System
        ↓
    Monitoring
        ↓
 ┌──────┼────────┐
 ↓      ↓        ↓
Errors  Speed   Accuracy
```

The FDE may monitor:

* **Errors** — Is the system failing?
* **Speed** — Is the system responding quickly?
* **API failures** — Are connections failing?
* **Model performance** — Is the AI giving good answers?
* **User feedback** — Are users satisfied?
* **Costs** — Is the system becoming too expensive?
* **Usage** — How many people are using it?
* **Data quality** — Is the data still reliable?

If something goes wrong, the FDE **finds the problem, fixes it, and improves the system**.


# Step 10 — Feedback and Improvement

The work does not stop after deployment.

Once real users start using the system, they provide **feedback**.

For example:

> **"The AI sometimes gives incorrect answers."**

The FDE investigates where the problem is coming from.

For example:

```text
Problem
   ↓
Poor Retrieval
   ↓
Wrong Documents
   ↓
Wrong Information
   ↓
Wrong Answer
```

The FDE may improve:

* **Chunking**
* **Embeddings**
* **Retrieval**
* **Prompts**
* **Model selection**
* **Evaluation**
* **Data quality**

Then the system is **tested again**.

This makes FDE work an **iterative process**:

```text
Build
  ↓
Deploy
  ↓
Measure
  ↓
Learn
  ↓
Improve
  ↓
Deploy Again
```


# FDE and AI

FDEs have become especially important with **AI and Generative AI**.

An AI model may be powerful, but every company has different:

* Data
* Business problems
* Workflows
* Security requirements
* Applications
* Users

The FDE connects the AI model to the customer's **real business environment**.

For example:

```text
General LLM
     ↓
Customer's Private Data
     ↓
    RAG
     ↓
Customer APIs
     ↓
Business Workflow
     ↓
Production AI Application
```

An FDE working with AI may work with:

* **LLMs**
* **RAG**
* **Prompt Engineering**
* **Embeddings**
* **Vector Databases**
* **APIs**
* **Model Evaluation**
* **AI Agents**
* **Multimodal AI**
* **Cloud Deployment**


# FDE vs Software Engineer

This is an important comparison because an **FDE is still a software engineer**, but their day-to-day work is different.

| Software Engineer                         | Forward Deployed Engineer                 |
| ----------------------------------------- | ----------------------------------------- |
| Builds the core product                   | Applies technology to customer problems   |
| Often works internally                    | Often works directly with customers       |
| Focuses on reusable features              | Focuses on customer-specific solutions    |
| Usually works in a controlled environment | Often works with messy real-world systems |
| Product-focused                           | Customer and outcome-focused              |
| Usually has less customer interaction     | Has high customer interaction             |


# FDE vs Machine Learning Engineer

A **Machine Learning Engineer** mainly focuses on **building, improving, and deploying ML models**.

A **Forward Deployed Engineer (FDE)** focuses on **using AI/ML technology to solve a specific customer's problem**.

### Simple Difference

> **ML Engineer:** "How can I make the model better?"

> **FDE:** "How can I make this technology work for this customer?"

### Example

An ML Engineer may build and optimize a recommendation model.

An FDE may take that model and connect it to the customer's **data, applications, APIs, and workflow**.

**ML Engineer → Model-focused**

**FDE → Customer and solution-focused**


# Skills Required for an FDE

An FDE needs a combination of **technical skills and soft skills** because they work directly with customers and their technology.

### Technical Skills

- Programming: Python, JavaScript, SQL
- Software Engineering: APIs, Git, Testing, Debugging
- Data: Databases, SQL, Data Integration
- AI/ML: Machine Learning, LLMs, RAG
- Deployment: Cloud, Docker, Monitoring

### Soft Skills

- Problem Solving
- Communication
- Business Understanding
- Customer Understanding

### FDE Mindset

An FDE should not immediately start coding when a customer asks for an AI solution.

They first ask:

1. What problem are we solving?
2. Who will use it?
3. What data do we need?
4. What systems do we need to connect?
5. How will we measure success?

> **FDE = Technical Skills + Problem Solving + Communication + Customer Understanding**

# Real-World Example: AI Assistant for Company Documents

Imagine a company has thousands of internal documents, and employees spend too much time searching for information.

The company wants an **AI assistant** that can quickly answer questions using those documents.

## Step 1 — Understand the Problem

The FDE first understands the problem:

> Employees are spending too much time finding information.

## Step 2 — Understand the Data and Environment

The FDE looks at where the company's information is stored:

* PDFs
* Word documents
* Web pages
* Databases

They also understand the company's existing systems and security requirements.

## Step 3 — Design and Build the Solution

The FDE designs a solution using technologies such as **RAG and an LLM**.

```text
Company Documents
        ↓
       RAG
        ↓
       LLM
        ↓
   AI Assistant
```

The FDE builds a small prototype and tests it with real company documents.

## Step 4 — Test and Integrate

The FDE tests the assistant using real employee questions.

They check:

* Are the answers correct?
* Is the information relevant?
* Does the AI hallucinate?
* Is it fast enough?

Then they connect it to the company's existing systems and authentication.

## Step 5 — Deploy and Improve

The solution is deployed into the company's environment.

The FDE monitors its performance, collects user feedback, and continuously improves the system.

> **This example shows what an FDE does: understand the customer's problem, work with their data and environment, build a solution, integrate it, test it, and make sure it works in the real world.**

### Simple FDE Flow

```text
Problem
   ↓
Data & Environment
   ↓
Build
   ↓
Test & Integrate
   ↓
Deploy & Improve
```
