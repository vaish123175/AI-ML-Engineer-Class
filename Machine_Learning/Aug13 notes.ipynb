## Ensemble Learning, KNN, SVM, and Clustering

## 1. Random Forest

Random Forest is a **supervised machine learning algorithm** that combines multiple decision trees to make a stronger and more stable prediction. Instead of depending on one decision tree, Random Forest creates many decision trees and combines their predictions. For classification, the final prediction is generally based on **majority voting**, while for regression, predictions are typically averaged.

A single decision tree can easily become very complex and fit the training data too closely. This creates an **overfitting problem**, where the model performs very well on training data but poorly on unseen data. Random Forest reduces this problem by combining multiple trees and averaging their results. This reduces the **variance** of the overall model.

### Why Random Forest is used

The main idea behind Random Forest is:

> **Multiple weak or unstable decision trees can be combined to create a more stable and accurate model.**

Random Forest uses **bagging (Bootstrap Aggregating)**. Different samples of the training data are created using random sampling with replacement, and each decision tree is trained on a different sample. The trees are also given random subsets of features when making splits.

Because the trees are trained independently, they can be trained **in parallel**.


# 2. Ensemble Learning

**Ensemble learning** is a machine learning technique where multiple models are combined to produce a better overall model.

Instead of depending on one model, ensemble methods use several models and combine their predictions.

The main advantages are:

* Reduce variance
* Reduce overfitting
* Improve prediction accuracy
* Make predictions more stable
* Combine the strengths of multiple models

Two major approaches to ensemble learning are:

1. **Bagging**
2. **Boosting**


# 3. Bagging

**Bagging** stands for **Bootstrap Aggregating**.

The main purpose of bagging is to reduce **variance**.

In bagging, multiple models are trained **independently and in parallel** using different samples of the training dataset.

For example, suppose we have one large dataset. Bagging creates several bootstrap samples from that dataset. Each sample is given to a different decision tree.

The process looks like:

```text
                 Training Data
                      |
        --------------------------------
        |              |               |
      Sample 1       Sample 2        Sample 3
        |              |               |
      Tree 1         Tree 2          Tree 3
        |              |               |
        --------------------------------
                      |
                Combine Results
                      |
                Final Prediction
```

Random Forest is one of the most important examples of a **bagging-based algorithm**.

### Why Bagging reduces variance

A single decision tree may make a very different prediction if the training data changes slightly. This means the tree has high variance.

When we combine many trees, individual errors tend to balance each other out. Therefore, the final model becomes more stable.



# 4. Boosting

**Boosting** is another ensemble learning technique.

Unlike bagging, where models are generally trained independently and in parallel, boosting trains models **sequentially**.

In boosting, the next model focuses on the mistakes made by the previous model.

The basic idea is:

```text
Training Data
     |
     v
  Model 1
     |
  Mistakes
     |
     v
  Model 2
     |
  Mistakes
     |
     v
  Model 3
     |
     v
Final Prediction
```

Each new model tries to improve the performance of the previous models.

Therefore:

* **Bagging → parallel**
* **Boosting → sequential**

Examples of boosting algorithms include:

* AdaBoost
* Gradient Boosting
* XGBoost
* LightGBM
* CatBoost


# 5. AdaBoost — Adaptive Boosting

**AdaBoost** stands for **Adaptive Boosting**.

AdaBoost is a boosting algorithm that combines several weak learners, usually small decision trees called **decision stumps**.

The word "adaptive" means that the algorithm changes the importance or **weights** of training examples based on previous mistakes.

Initially, all training observations have equal weights.

The first model is trained. After the model makes predictions, incorrectly classified observations receive more importance. The next model pays more attention to these difficult observations.

This process continues for multiple models.

### AdaBoost process

```text
Initial Data
     |
Equal Weights
     |
     v
 Model 1
     |
Identify Mistakes
     |
Increase Weights of Incorrect Examples
     |
     v
 Model 2
     |
Identify Mistakes
     |
Adjust Weights Again
     |
     v
 Model 3
     |
     v
Final Combined Prediction
```

The predictions of all the weak learners are then combined to produce the final prediction.

### Important point

In AdaBoost, the weights are updated according to the performance of previous models. The algorithm therefore focuses progressively more on difficult observations.

# 6. Gradient Boosting

**Gradient Boosting** is another sequential ensemble learning algorithm.

Instead of explicitly increasing the weights of incorrectly classified observations in the same way as AdaBoost, Gradient Boosting builds new models to reduce the **loss function** or prediction error of the existing ensemble.

The first model makes predictions and produces errors.

The next model focuses on these errors and tries to correct them.

This continues sequentially.

```text
Model 1
   |
Calculate Error
   |
Model 2 learns to reduce Error
   |
Calculate Remaining Error
   |
Model 3 learns to reduce Remaining Error
   |
   v
Final Model
```

The word **gradient** comes from the use of gradient-based optimization to minimize the loss function.

### AdaBoost vs Gradient Boosting

| AdaBoost                                       | Gradient Boosting                     |
| ---------------------------------------------- | ------------------------------------- |
| Adjusts weights of observations                | Optimizes a loss function             |
| Focuses on incorrectly classified observations | Focuses on reducing prediction errors |
| Sequential                                     | Sequential                            |
| Uses weak learners                             | Usually uses decision trees           |
| Based on adaptive weighting                    | Based on gradient-based optimization  |


# 7. XGBoost — Extreme Gradient Boosting

**XGBoost** stands for **Extreme Gradient Boosting**.

It is an advanced and highly optimized implementation of gradient boosting. XGBoost became popular because it provides strong predictive performance while also incorporating several techniques to control overfitting and improve computational efficiency.

XGBoost uses regularization techniques such as:

* **L1 regularization**
* **L2 regularization**

Regularization helps prevent the model from becoming unnecessarily complex and reduces overfitting.

XGBoost can also handle missing values efficiently and has mechanisms that make it suitable for sparse datasets.

It also supports features such as:

* Regularization
* Parallel computation
* Missing-value handling
* Cross-validation
* Early stopping
* Efficient tree construction
* Handling of sparse data

### Why XGBoost is popular

XGBoost is commonly used when we need a powerful model for structured or tabular data. It often performs very well on classification and regression problems.


# 8. CatBoost — Categorical Boosting

**CatBoost** stands for **Categorical Boosting**.

CatBoost is a gradient boosting algorithm designed particularly to work well with **categorical features**.

Categorical data does not necessarily mean only text. Examples include:

* Gender
* City
* Product category
* Department
* Customer type
* Payment method

Traditional machine learning algorithms usually require categorical variables to be converted into numerical representations before training.

CatBoost provides specialized techniques for handling categorical features directly and can reduce the need for extensive preprocessing.

Therefore:

> **CatBoost is especially useful when a dataset contains many categorical features.**


# 9. LightGBM

**LightGBM**, developed by Microsoft, is another gradient boosting framework designed for high-performance machine learning.

It is particularly useful when dealing with **large datasets and high-dimensional data**.

One important difference is how LightGBM grows its decision trees.

Traditional tree algorithms may use **level-wise growth**, where nodes at the same level are expanded.

LightGBM uses **leaf-wise tree growth**. It chooses the leaf that provides the largest reduction in the loss and splits that leaf.

```text
Level-wise:
        Root
       /    \
      /      \
   Split    Split
   both branches at the same level


Leaf-wise:
        Root
       /    \
      /      \
    Leaf    Leaf
              |
            Split
              |
            Split
```

Leaf-wise growth can provide better loss reduction and can make LightGBM very efficient on large datasets. However, without appropriate controls, leaf-wise growth can also increase the risk of overfitting.

LightGBM is commonly useful for:

* Large datasets
* High-dimensional datasets
* Fast model training
* Classification
* Regression
* Ranking problems

# 11. Overfitting, Variance, and Bias

Many ensemble techniques are designed to improve the balance between **bias and variance**.

### Variance

Variance describes how much a model's prediction changes when the training data changes.

A decision tree can have high variance because a small change in the training data can produce a very different tree.

Random Forest and bagging are especially useful for reducing variance.

### Bias

Bias occurs when a model is too simple to capture the underlying pattern in the data.

A model with high bias tends to **underfit**.

### Overfitting

Overfitting occurs when the model learns the training data too closely, including noise and unnecessary patterns.

The model performs well on training data but poorly on unseen data.

### Underfitting

Underfitting occurs when the model is too simple and fails to learn important patterns from the data.

The goal of machine learning is to build a model that generalizes well to unseen data.


# 12. Class Imbalance

**Class imbalance** occurs when one class contains significantly more observations than another class.

For example:

```text
Normal Transactions = 9,500
Fraud Transactions  =   500
```

Here, the dataset is imbalanced because the fraud class represents a much smaller portion of the data.

A model could achieve high accuracy simply by predicting the majority class most of the time, while performing poorly on the minority class.

Class imbalance can be addressed using techniques such as:

* Class weights
* Oversampling
* Undersampling
* SMOTE
* Appropriate evaluation metrics
* Algorithm-specific imbalance handling

Metrics such as **precision, recall, F1-score, and ROC-AUC** can be more informative than accuracy when classes are highly imbalanced.

# 13. K-Nearest Neighbors — KNN

**K-Nearest Neighbors (KNN)** is a **supervised machine learning algorithm**.

KNN is called a **distance-based algorithm** because it makes predictions based on the distance between observations.

The basic idea is:

> An observation is likely to belong to the same class as the observations that are closest to it.

Suppose we want to classify a new customer. KNN calculates the distance between the new customer and existing customers.

It then finds the **K nearest observations**.

For classification, the majority class among those neighbors is generally selected as the prediction.


# 14. K Value in KNN

The **K value** represents how many nearest neighbors the algorithm should consider.

For example, if:

```text
K = 3
```

the algorithm considers the three closest observations.

If two of the three neighbors belong to Class A and one belongs to Class B, the new observation is classified as Class A.

```text
K = 5

Class A → 3 neighbors
Class B → 2 neighbors

Prediction → Class A
```

Choosing the correct K value is important.

A very small K can make the model sensitive to noise and can lead to overfitting.

A very large K can make the model too generalized and may lead to underfitting.


# 15. Euclidean Distance

One common distance measure used in KNN is **Euclidean distance**.

For two points:

```text
A = (x1, y1)
B = (x2, y2)
```

the Euclidean distance is:

```text
Distance = √((x2 - x1)² + (y2 - y1)²)
```

The smaller the distance, the closer the observations are.

Because KNN depends on distance, **feature scaling is very important**. If one feature has values ranging from 0–100000 and another ranges from 0–10, the larger-scale feature can dominate the distance calculation.

Common scaling techniques include:

* Standardization
* Min-Max scaling

# 16. KNN for Classification and Regression

KNN can be used for both classification and regression.

### Classification

The model uses the majority class among the K nearest neighbors.

### Regression

The model generally uses the average value of the K nearest neighbors.

Therefore:

> KNN is a supervised algorithm that can solve both classification and regression problems.


# 17. Support Vector Machine — SVM

**SVM**, or **Support Vector Machine**, is a supervised machine learning algorithm used primarily for classification but also available for regression.

The main objective of SVM is to find the best **hyperplane** that separates different classes.

For a two-dimensional dataset, the hyperplane is a line.

For higher-dimensional data, it is called a hyperplane.

```text
Class A       |       Class B
   ● ● ●      |      ○ ○ ○
   ● ● ●      |      ○ ○ ○
               ^
           Hyperplane
```

The SVM does not simply try to find any line that separates the classes. It tries to find the separating boundary with the **maximum margin**.

# 18. Margin in SVM

The **margin** is the distance between the decision boundary and the closest observations from each class.

The closest observations are called **support vectors**.

The objective is generally:

> Find the hyperplane that maximizes the margin between the classes.

A larger margin generally provides better generalization.

```text
Class A      Support Vector       Support Vector      Class B
   ●                 ●                  ○                 ○
   ●                 ●       |          ○                 ○
   ●                         |                            ○
                             |
                         Hyperplane
```

The observations closest to the hyperplane are the most important points for defining the decision boundary.


# 19. Non-Linear SVM

Sometimes the data cannot be separated using a straight line.

For example:

```text
    ○ ○ ○
   ○ ● ● ○
   ○ ● ● ○
    ○ ○ ○
```

A straight line cannot properly separate the classes.

SVM can handle nonlinear relationships using **kernel functions**.

Common kernels include:

* Linear kernel
* Polynomial kernel
* RBF (Radial Basis Function)
* Sigmoid kernel

The **kernel trick** allows SVM to effectively work in a higher-dimensional feature space without explicitly calculating all the transformed features.

# 20. SVM for Regression

SVM can also be used for regression. The regression version is called **Support Vector Regression (SVR)**.

Instead of finding a classification boundary, SVR attempts to find a function that fits the data while allowing a certain amount of error.

Therefore:

* **SVC → Support Vector Classification**
* **SVR → Support Vector Regression**

# 22. Unsupervised Learning

**Unsupervised learning** is different from supervised learning because the dataset does not have a labeled target variable.

The model tries to discover hidden patterns, groups, or structures within the data.

The major clustering algorithms covered are:

1. K-Means Clustering
2. DBSCAN
3. Hierarchical Clustering

# 23. K-Means Clustering

**K-Means** is an unsupervised clustering algorithm.

The objective is to divide observations into **K clusters**.

The algorithm generally follows these steps:

1. Select the number of clusters, K.
2. Initialize cluster centers.
3. Assign each observation to the closest cluster center.
4. Recalculate the cluster centers.
5. Repeat the assignment and recalculation process.
6. Stop when the cluster assignments or centers stabilize.

For example, customer data can be divided into groups such as:

```text
Cluster 1 → Low-value customers
Cluster 2 → Medium-value customers
Cluster 3 → High-value customers
```

K-Means is distance-based, so feature scaling can be important.

# 24. DBSCAN

**DBSCAN** stands for **Density-Based Spatial Clustering of Applications with Noise**.

Unlike K-Means, DBSCAN does not require us to specify the number of clusters beforehand.

It groups observations based on **density**.

DBSCAN can identify:

* Dense regions
* Noise
* Outliers
* Irregularly shaped clusters

This makes DBSCAN useful when the clusters are not naturally circular or spherical.

Two important parameters are:

* **Epsilon (eps)** — defines the neighborhood distance.
* **MinPts** — minimum number of observations required to form a dense region.

DBSCAN is particularly useful when the dataset contains noise or outliers.

# 25. Hierarchical Clustering

**Hierarchical clustering** creates a hierarchy of clusters.

There are two major approaches:

### Agglomerative clustering

This is a bottom-up approach.

Initially, every observation is treated as its own cluster. The closest clusters are then repeatedly merged.

```text
A   B   C   D
|   |   |   |
 \ /     \ /
 AB      CD
    \   /
     ABCD
```

### Divisive clustering

This is a top-down approach.

Initially, all observations belong to one cluster, and the cluster is repeatedly divided into smaller clusters.

The resulting hierarchy can be visualized using a **dendrogram**.
