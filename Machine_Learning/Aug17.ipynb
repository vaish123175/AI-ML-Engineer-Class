# 3. Tokenization

**Tokenization** is the process of breaking text into smaller units called **tokens**.

A token can be:

* A sentence
* A word
* A subword
* Sometimes a character

For example:

```text
"I love machine learning."
```

After word tokenization:

```text
["I", "love", "machine", "learning", "."]
```

Tokenization is one of the first and most important steps in NLP because machine learning models generally cannot directly work with raw paragraphs.

## 3.1 Sentence Tokenization

**Sentence tokenization** divides a paragraph or document into individual sentences.

Example:

```text
"Python is easy to learn. NLP is interesting. Machine learning is powerful."
```

After sentence tokenization:

```text
[
    "Python is easy to learn.",
    "NLP is interesting.",
    "Machine learning is powerful."
]
```

Sentence tokenization is useful when we need to process each sentence separately.

## 3.2 Word Tokenization

**Word tokenization** breaks a sentence into individual words and tokens.

Example:

```text
"Python is easy to learn."
```

Output:

```text
["Python", "is", "easy", "to", "learn", "."]
```

# 5. Stop Words

**Stop words** are commonly occurring words that may provide little useful information for certain NLP tasks.

Examples include:

```text
the
is
a
an
and
or
of
to
in
on
for
```

For example:

```text
"The cat is sitting on the table."
```

After removing stop words:

```text
"cat sitting table"
```

The purpose of removing stop words is to reduce unnecessary information and sometimes reduce the number of features in a model.

### Important

Stop words should **not always be removed**.

For example, in sentiment analysis:

```text
"I am not happy"
```

The word **"not"** is extremely important because removing it can change the meaning.


# 6. POS Tagging

**POS stands for Part of Speech.**

**POS tagging** assigns a grammatical category to each word in a sentence.

Common POS categories include:

| POS         | Meaning                    | Example   |
| ----------- | -------------------------- | --------- |
| Noun        | Person, place, thing       | student   |
| Verb        | Action                     | run       |
| Adjective   | Describes a noun           | beautiful |
| Adverb      | Describes a verb/adjective | quickly   |
| Pronoun     | Replaces a noun            | he        |
| Preposition | Shows relationship         | in        |
| Conjunction | Connects words             | and       |
| Determiner  | Introduces a noun          | the       |

Example:

```text
"Vaishnavi studies Python."
```

Possible POS tags:

```text
Vaishnavi → Proper Noun
studies → Verb
Python → Proper Noun
```

POS tagging helps NLP systems understand the grammatical structure of a sentence.

# 7. Chunking

**Chunking** is the process of grouping words into meaningful phrases based on their POS tags.

For example:

```text
"The beautiful girl is reading a book."
```

Possible chunks:

```text
[The beautiful girl] → Noun Phrase

[is reading] → Verb Phrase

[a book] → Noun Phrase
```

Chunking is also called **shallow parsing**.

It is useful for identifying phrases such as:

* Noun phrases
* Verb phrases
* Prepositional phrases

# 8. Stemming

**Stemming** is the process of reducing a word to its root or base-like form by removing prefixes or suffixes.

Examples:

```text
playing → play
played → play
plays → play
```

Another example:

```text
studies → studi
```

The important point is that stemming does not always produce a proper English word.

### Example

```text
"connected"
"connecting"
"connection"
```

A stemmer may produce something like:

```text
connect
```

or another shortened form depending on the algorithm.

# 9. Lemmatization

**Lemmatization** also reduces words to their base form, but it tries to produce a **valid dictionary word**.

Example:

```text
better → good
running → run
studies → study
cars → car
```

Lemmatization generally uses linguistic information to determine the correct base form.


# 10. Stemming vs Lemmatization

| Stemming                  | Lemmatization                  |
| ------------------------- | ------------------------------ |
| Removes prefixes/suffixes | Finds the base dictionary form |
| Faster                    | Usually slower                 |
| Less accurate             | More accurate                  |
| May produce invalid words | Usually produces valid words   |
| Uses simple rules         | Uses linguistic information    |

### Example

```text
Studies
```

Stemming might produce:

```text
studi
```

Lemmatization produces:

```text
study
```

# 11. Drawbacks of Stemming

Stemming is fast, but it has several disadvantages.

### 1. It can produce invalid words

Example:

```text
studies → studi
```

`studi` is not a proper English word.

### 2. Over-stemming

Different words may incorrectly be reduced to the same stem.

### 3. Under-stemming

Words that should have the same root may sometimes remain different.

### 4. It does not understand context

A stemmer generally works using rules rather than understanding the meaning of the sentence.

# 13. BIO Tagging

**BIO tagging** is a method commonly used for identifying entities or chunks in a sentence.

BIO stands for:

* **B = Beginning**
* **I = Inside**
* **O = Outside**

Example:

```text
Elon Musk works at SpaceX.
```

BIO tagging could be:

```text
Elon → B-PERSON
Musk → I-PERSON
works → O
at → O
SpaceX → B-ORGANIZATION
```

### Meaning

**B-PERSON**

The word is the beginning of a person entity.

**I-PERSON**

The word is inside an already-started person entity.

**O**

The word does not belong to an entity.

BIO tagging is particularly useful when an entity contains multiple words.

**Vectorization** is the process of converting text into numerical representations.

# 15. One-Hot Encoding

**One-hot encoding** represents each word using a binary vector.

Suppose our vocabulary is:

```text
["cat", "dog", "fish"]
```

We can represent the words as:

```text
cat  → [1, 0, 0]
dog  → [0, 1, 0]
fish → [0, 0, 1]
```

Only one position contains `1`, while all other positions contain `0`.

### Advantages

* Simple
* Easy to understand
* Works well for small vocabularies

### Disadvantages

* Creates very large vectors for large vocabularies
* Sparse representation
* Does not capture meaning
* Does not understand relationships between words

For example:

```text
king
queen
```

One-hot encoding does not know that these words are semantically related.


# 16. Bag of Words (BoW)

**Bag of Words** is a text representation technique that represents a document based on the words it contains and their frequency.

It ignores:

* Word order
* Grammar
* Context

Example documents:

```text
Document 1: "I love Python"
Document 2: "I love NLP"
```

Vocabulary:

```text
["I", "love", "Python", "NLP"]
```

Representation:

```text
Document 1 → [1, 1, 1, 0]
Document 2 → [1, 1, 0, 1]
```

The model focuses on which words occur rather than their order.


# 17. Count Vectorizer

**CountVectorizer** is a technique provided by Scikit-learn that converts text documents into numerical vectors based on word counts.


### Important

CountVectorizer is essentially an implementation of the **Bag of Words** approach.


# 18. TF-IDF

**TF-IDF** stands for:

> **Term Frequency - Inverse Document Frequency**

It is a technique used to determine how important a word is within a document relative to a collection of documents.

A word gets a high TF-IDF score when:

* It appears frequently in a particular document.
* It does not appear frequently across many other documents.

A word gets a lower score when it occurs in many documents.

## 18.1 Term Frequency (TF)

**Term Frequency** measures how frequently a word appears in a document.

A simple formula is:

```text
TF = Number of times the term appears in the document
     -----------------------------------------------
     Total number of terms in the document
```

For example, if the word `Python` appears 3 times in a document containing 100 words:

```text
TF = 3 / 100
   = 0.03
```


## 18.2 Inverse Document Frequency (IDF)

**IDF** measures how rare or common a word is across all documents.

A simplified formula is:

```text
IDF = log(Total number of documents / Number of documents containing the term)
```

If a word appears in almost every document, its IDF is low.

If a word appears in only a few documents, its IDF is high.


## 18.3 TF-IDF Formula

```text
TF-IDF = TF × IDF
```

Therefore:

```text
High TF-IDF
    ↓
Important for this document
    ↓
Not common across all documents
```

# 19. N-Grams

An **N-gram** is a sequence of `N` consecutive words or tokens.

N-grams are useful because they can capture relationships between neighboring words.


## 19.1 Unigram

A **unigram** contains one word.

Example:

```text
"I love machine learning"
```

Unigrams:

```text
I
love
machine
learning
```


## 19.2 Bigram

A **bigram** contains two consecutive words.

Example:

```text
"I love machine learning"
```

Bigrams:

```text
I love
love machine
machine learning
```

## 19.3 Trigram

A **trigram** contains three consecutive words.

Example:

```text
"I love machine learning"
```

Trigrams:

```text
I love machine
love machine learning
```

## 19.4 N-Gram

An N-gram can contain any number of consecutive tokens.

```text
1 word  → Unigram
2 words → Bigram
3 words → Trigram
N words → N-gram
```


# 20. Skip-Grams

A **Skip-gram** allows words to be skipped when creating word relationships.

Instead of considering only adjacent words, it can learn relationships between words within a particular window.

Example:

```text
"The cat is sitting"
```

A skip-gram approach can learn relationships such as:

```text
cat → sitting
cat → is
```

depending on the selected window size.

Skip-gram is also an important architecture used in **Word2Vec**.


# 21. Word Embeddings

**Word embedding** is a technique for representing words as dense numerical vectors.

Unlike one-hot encoding, word embeddings can capture relationships between words.

For example:

```text
king
queen
man
woman
```

can be represented using vectors that capture semantic relationships.

Conceptually:

```text
king → [0.25, 0.71, -0.32, ...]
queen → [0.28, 0.69, -0.29, ...]
```

Words with similar meanings tend to have similar vector representations.

### Examples of Word Embedding Techniques

* Word2Vec
* GloVe
* FastText
* Transformer embeddings


# 22. Continuous Bag of Words (CBOW)

**CBOW stands for Continuous Bag of Words.**

It is one of the architectures used by **Word2Vec**.

CBOW tries to predict a target word based on its surrounding context words.

Example:

```text
"The cat is ___ on the mat."
```

The model uses surrounding words to predict:

```text
sitting
```

Conceptually:

```text
Context Words
      ↓
    CBOW
      ↓
Target Word
```

CBOW is generally efficient and works well when there is a lot of training data.


# 23. RNN

**RNN stands for Recurrent Neural Network.**

RNNs are neural networks designed to process sequential data.

They are useful for data where order matters, such as:

* Text
* Speech
* Time series
* Sentences

Example:

```text
"I love machine learning"
```

The model processes the sequence step by step:

```text
I
 ↓
love
 ↓
machine
 ↓
learning
```

The previous information is carried forward to help understand the next input.


## 23.1 RNN Architecture

Conceptually:

```text
Input 1 → RNN → Hidden State
              ↓
Input 2 → RNN → Hidden State
              ↓
Input 3 → RNN → Hidden State
              ↓
Input 4 → RNN → Output
```

The hidden state acts as a form of memory.


# 24. Problems with Basic RNNs

Basic RNNs have difficulty remembering information over long sequences.

One major problem is the:

**Vanishing Gradient Problem**

During training, gradients can become extremely small as they are propagated backward through many time steps.

As a result, the model can have difficulty learning long-term dependencies.

For example:

```text
"The student who studied very hard for the exam eventually..."
```

The important information may be many words earlier in the sentence.

A basic RNN may struggle to retain that information.

This led to more advanced architectures such as **LSTM** and **GRU**.


# 25. LSTM

**LSTM stands for Long Short-Term Memory.**

LSTM is a type of recurrent neural network designed to handle long-term dependencies better than a basic RNN.

LSTM uses gates to control what information should be:

* Forgotten
* Stored
* Updated
* Passed forward

The main components are:

### Forget Gate

Determines which information should be removed from memory.

### Input Gate

Determines which new information should be stored.

### Output Gate

Determines which information should be sent to the next step.

Conceptually:

```text
Input
  ↓
┌─────────────────────┐
│        LSTM         │
│                     │
│  Forget Gate        │
│  Input Gate         │
│  Output Gate        │
│                     │
└─────────────────────┘
  ↓
Output
```


# 26. Drawbacks of LSTM

Although LSTM improves upon basic RNNs, it has some disadvantages.

### 1. Computationally expensive

LSTMs have multiple gates and therefore require more computation.

### 2. More parameters

The architecture contains many parameters that need to be trained.

### 3. Slower training

Because of the sequential nature of recurrent networks, training can be slower.

### 4. Difficult to parallelize

Unlike Transformers, RNN-based architectures process sequences sequentially.

### 5. Still has limitations with very long sequences

LSTMs improve long-term memory but do not completely eliminate the challenges of handling very long sequences.


# 27. GRU

**GRU stands for Gated Recurrent Unit.**

GRU is another type of recurrent neural network designed to handle long-term dependencies.

It is similar to LSTM but has a simpler architecture.

GRU mainly uses:

* Update Gate
* Reset Gate

### GRU vs LSTM

| LSTM                           | GRU                                 |
| ------------------------------ | ----------------------------------- |
| More complex                   | Simpler                             |
| Uses more gates                | Uses fewer gates                    |
| More parameters                | Fewer parameters                    |
| More computationally expensive | Generally faster                    |
| Separate cell state            | Does not have a separate cell state |

GRUs can perform similarly to LSTMs on many tasks while being computationally simpler.


# 28. Seq2Seq

**Seq2Seq stands for Sequence-to-Sequence.**

A Seq2Seq model takes one sequence as input and produces another sequence as output.

It is commonly associated with an:

* **Encoder**
* **Decoder**

Architecture:

```text
Input Sequence
      ↓
   Encoder
      ↓
Context Representation
      ↓
   Decoder
      ↓
Output Sequence
```


# 29. Encoder

The **encoder** reads the input sequence and converts it into a representation containing information about the input.

Example:

```text
Input:
"I love Python"
       ↓
    Encoder
       ↓
Context Representation
```

The context representation is then passed to the decoder.



# 30. Decoder

The **decoder** uses the encoded representation to generate the output sequence.

Example:

```text
Input:
"I love Python"
       ↓
    Encoder
       ↓
Representation
       ↓
    Decoder
       ↓
"J'aime Python"
```

This type of architecture was widely used for tasks such as:

* Machine translation
* Text summarization
* Conversational systems



# 31. Transformer Model

The **Transformer** is a neural network architecture designed primarily for processing sequences using **attention mechanisms** rather than relying entirely on recurrent processing.

The Transformer architecture was introduced in the paper:

> **"Attention Is All You Need"**

Transformers became extremely important in modern NLP.

They are the foundation of many modern language models.


# 32. Transformer Architecture

The original Transformer consists of two major components:

```text
              TRANSFORMER
                  │
        ┌─────────┴─────────┐
        ↓                   ↓
     Encoder             Decoder
        ↓                   ↓
  Self-Attention       Masked Attention
        ↓                   ↓
  Feed Forward         Cross Attention
        ↓                   ↓
        └─────────┬─────────┘
                  ↓
                Output
```

The architecture uses **attention**, particularly **multi-head attention**, to determine which parts of the input are important when processing each token.


# 33. Attention Mechanism

**Attention** allows a model to focus on different words in a sequence when processing a particular word.

Example:

```text
"The animal didn't cross the road because it was tired."
```

When understanding the word:

```text
"it"
```

the model needs to determine what `it` refers to.

Attention helps the model identify relevant relationships between words.


# 34. Multi-Head Attention

**Multi-Head Attention** allows the model to perform multiple attention operations in parallel.

Instead of having only one attention mechanism:

```text
Input
  ↓
Attention
  ↓
Output
```

the Transformer uses multiple attention heads:

```text
             Input
               ↓
     ┌─────────┼─────────┐
     ↓         ↓         ↓
  Head 1     Head 2    Head 3
     ↓         ↓         ↓
     └─────────┼─────────┘
               ↓
          Combined Output
```

Different attention heads can learn different relationships between words.

For example, one head might focus on:

* Grammatical relationships

Another might focus on:

* Subject-object relationships

Another might focus on:

* Long-distance relationships between words



# 35. Transformer vs RNN

| RNN                                       | Transformer                                         |
| ----------------------------------------- | --------------------------------------------------- |
| Processes sequence sequentially           | Can process many tokens in parallel during training |
| Uses recurrent connections                | Uses attention                                      |
| More difficult to parallelize             | Highly parallelizable                               |
| Can struggle with long-term dependencies  | Attention can directly connect distant tokens       |
| Generally slower for large-scale training | Efficient for large-scale training                  |
| Uses hidden states                        | Uses attention representations                      |
