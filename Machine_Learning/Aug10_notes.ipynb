# Logistic Regression

## 1. What is Logistic Regression?

Logistic Regression is a **supervised machine learning algorithm used for classification problems**. Unlike Linear Regression, which predicts a continuous numerical value, Logistic Regression predicts the probability that an observation belongs to a particular class. For example, it can predict whether a student will **Pass or Fail**, whether a customer will **Buy or Not Buy**, or whether an email is **Spam or Not Spam**.

Logistic Regression is called regression because it calculates a numerical probability, but its main purpose is **classification**.

It is used when the target/output variable is **categorical** rather than continuous.

For example:

- Pass / Fail
- Yes / No
- Spam / Not Spam
- Disease / No Disease
- Customer will buy / Customer will not buy

Although it is called "regression", Logistic Regression is mainly used for **classification problems**.

The model calculates a probability and then uses a **threshold value** to decide which class the observation belongs to.

### Example

Suppose we want to predict whether a student will pass an exam.

Input:
- Hours studied
- Attendance
- Previous marks

Output:
- 0 = Fail
- 1 = Pass

The model may produce:

```text
Probability = 0.85

If the threshold is 0.5:


0.85 >= 0.5 → Pass (Class 1)


#  Linear Regression vs Logistic Regression

| Linear Regression | Logistic Regression |
|---|---|
| Used for regression problems | Used for classification problems |
| Predicts continuous values | Predicts class probabilities |
| Output can be any continuous number | Output is between 0 and 1 |
| Uses a linear equation | Uses a linear equation followed by a sigmoid function |
| Example: Predict house price | Example: Predict Pass/Fail |
| Evaluation: MSE, RMSE, R² | Evaluation: Accuracy, Precision, Recall, F1, ROC-AUC |

## Linear Regression

Linear Regression predicts a **continuous numerical value**.

### Example

```text
Hours Studied → Exam Score

5 hours → 72 marks
7 hours → 84 marks
9 hours → 91 marks



## Logistic Regression Equation

Logistic Regression first calculates a linear combination of the input features:

`z = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ`

The value `z` can be any number, including negative or positive values. This value is then passed through the **Sigmoid Function** to convert it into a probability between 0 and 1.


# 2. Linear Equation in Logistic Regression

Logistic Regression uses a **linear equation** to calculate a value called `z`.

The general equation is:

$$
z = b_0 + b_1x_1 + b_2x_2 + ... + b_nx_n
$$

Where:

- `z` = output of the linear equation
- `b₀` = intercept
- `b₁, b₂, ..., bₙ` = coefficients
- `x₁, x₂, ..., xₙ` = input features

The value of `z` can be any numerical value.

It can be:

```text
-10
-5
-2
0
2
5
10


However, Logistic Regression needs a probability between 0 and 1.

Therefore, the value of z is passed through the Sigmoid Function.

Process

Input Features
      ↓
Linear Equation
      ↓
Calculate z
      ↓
Sigmoid Function
      ↓
Probability between 0 and 1
      ↓
Threshold
      ↓
Class 0 or Class 1



# 3. Sigmoid Function

The **Sigmoid Function** is the main function used in binary Logistic Regression. It takes any value of `z` and converts it into a value between **0 and 1**.


### Formula

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

Where:

- `σ(z)` = output of the Sigmoid Function
- `z` = value calculated from the linear equation
- `e` = mathematical constant approximately equal to `2.718`

### How It Works

The linear equation can produce any value:

```text
z = -10
z = -5
z = 0
z = 5
z = 10

For example:

* Probability = 0.90 → very likely Class 1
* Probability = 0.70 → likely Class 1
* Probability = 0.40 → likely Class 0
* Probability = 0.10 → very likely Class 0

The sigmoid function is useful because classification models need a probability that can be converted into a class.

The Sigmoid Function converts these values into probabilities:

z → Sigmoid Function → Probability

-10 → 0.000

 -5 → 0.007

  0 → 0.500

  5 → 0.993
  
 10 → 1.000

 The output is always between 0 and 1.

# 4. Threshold Value

A **threshold value** is used in Logistic Regression to convert the predicted probability into a **class label**.

The Sigmoid Function gives us a probability between `0` and `1`, but we need to convert that probability into a class such as `0` or `1`.

The commonly used default threshold is:

```text
0.5

Classification Rule

Probability >= 0.5
        ↓
Class 1

Probability < 0.5
        ↓
Class 0


For example, if the model predicts a probability of 0.75, it is classified as Class 1. If the probability is 0.25, it is classified as Class 0.

The threshold does not always have to be 0.5. It can be changed depending on the business problem and the importance of precision or recall.

# 5. Cost Function

A **cost function** measures how far the model's predictions are from the actual class labels. Logistic Regression commonly uses **Log Loss**, also called **Binary Cross-Entropy Loss**, instead of the Mean Squared Error normally associated with Linear Regression.

The purpose of the cost function is to give a higher penalty when the model makes confident but incorrect predictions. During training, the model adjusts its parameters to minimize this cost.

A lower cost generally means that the model's predicted probabilities are closer to the actual outcomes.

It measures the difference between:

```text
Actual Class
     ↓
vs.
Predicted Probability

The goal of the model is to minimize the cost function.

Why Do We Need a Cost Function?


The cost function tells us how wrong the model's predictions are.

Low Cost
   ↓
Predictions are closer to the actual values
   ↓
Better model

High Cost
   ↓
Predictions are far from the actual values
   ↓
Poorer model


Example

Suppose the actual class is:

Actual = 1

If the model predicts:

Probability = 0.95

The model is very confident and correct.

Actual = 1

Predicted Probability = 0.95

        ↓
        
Low Cost

But if the model predicts:

Probability = 0.05

The model is very confident but wrong.

Actual = 1

Predicted Probability = 0.05

        ↓

High Cost

Therefore, Log Loss gives a large penalty to confident incorrect predictions.

# 6. Multiple Logistic Regression

**Multiple Logistic Regression** is used when we have **multiple input features** to predict a classification outcome.

Instead of using only one input variable, the model uses two or more input variables.

### Example

Suppose we want to predict whether a student will **Pass or Fail**.

We can use multiple input features:

```text
Hours Studied
Attendance
Previous Exam Score

The model combines all these features to calculate `z`, and the sigmoid function converts `z` into a probability.

So, Multiple Logistic Regression means **one classification target with multiple input features**.



## 7. Multinomial Logistic Regression

Binary Logistic Regression is generally used when there are **two classes**, such as:

* Pass / Fail
* Yes / No
* Spam / Not Spam

When there are **more than two classes**, Multinomial Logistic Regression can be used.

For example, suppose we want to classify a product into:

* Electronics
* Clothing
* Food

The model calculates probabilities for each class and selects the class with the highest probability.



## 8. Softmax Function

The **Softmax Function** is commonly used in Multinomial Logistic Regression to convert the model's output scores into probabilities.

Unlike the sigmoid function, which is generally used for binary classification, Softmax produces a probability for **each class**.

The probabilities produced by Softmax have two important properties:

1. Every probability is between 0 and 1.
2. The probabilities of all classes add up to 1.

For example:

| Class       | Probability |
| ----------- | ----------: |
| Electronics |        0.20 |
| Clothing    |        0.15 |
| Food        |        0.65 |

The sum is:

`0.20 + 0.15 + 0.65 = 1.00`

Since **Food** has the highest probability, the model predicts **Food** as the class label.

The model internally calculates scores from the input features. These scores are transformed by Softmax into probabilities.



## 9. Evaluation of Classification Models

After training a classification model, we need to determine **how well the model is performing**. Different evaluation metrics are used depending on the problem.

Common classification evaluation metrics include:

* Confusion Matrix
* Accuracy
* Precision
* Recall
* Specificity
* F1 Score
* ROC Curve
* AUC



## 10. Confusion Matrix

A **Confusion Matrix** is used to understand the different types of predictions made by a classification model.

It contains four important values:

### True Positive (TP)

The actual class is positive and the model correctly predicts positive.

Example: A student actually passes and the model predicts Pass.

### True Negative (TN)

The actual class is negative and the model correctly predicts negative.

Example: A student actually fails and the model predicts Fail.

### False Positive (FP)

The actual class is negative, but the model incorrectly predicts positive.

This is also called a **Type I Error**.

Example: A student actually fails, but the model predicts Pass.

### False Negative (FN)

The actual class is positive, but the model incorrectly predicts negative.

This is also called a **Type II Error**.

Example: A student actually passes, but the model predicts Fail.



## 11. Accuracy

**Accuracy** measures the percentage of total predictions that the model classified correctly.

Formula:

`Accuracy = (TP + TN) / (TP + TN + FP + FN)`

For example, if a model correctly predicts 90 out of 100 observations, its accuracy is 90%.

Accuracy is useful when the classes are reasonably balanced. However, accuracy can be misleading when the dataset has **imbalanced classes**.



## 12. Precision

**Precision** tells us, out of everything the model predicted as positive, how many were actually positive.

Formula:

`Precision = TP / (TP + FP)`

Precision is especially important when **False Positives are costly**.

For example, in a spam detection system, we may care about avoiding legitimate emails being incorrectly classified as spam.

A high precision means that when the model predicts positive, it is usually correct.



## 13. Recall or Sensitivity

**Recall**, also called **Sensitivity or True Positive Rate**, tells us how many of the actual positive cases the model successfully identified.

Formula:

`Recall = TP / (TP + FN)`

Recall is especially important when **False Negatives are costly**.

For example, if a model is being used to identify fraudulent transactions, missing an actual fraud case can be more serious than incorrectly flagging some legitimate transactions.

A high recall means that the model is good at finding actual positive cases.



## 14. Precision vs Recall Trade-Off

There is often a **trade-off between Precision and Recall**.

Increasing the threshold can make the model more selective about predicting positive cases. This can increase precision but may reduce recall.

Decreasing the threshold can make the model predict more positive cases. This can increase recall but may reduce precision.

Therefore, the best threshold depends on the business problem.

The goal is not always to maximize both independently. Instead, we choose a balance that is appropriate for the problem.



## 15. F1 Score

The **F1 Score** combines Precision and Recall into a single metric. It is the **harmonic mean** of Precision and Recall.

Formula:

`F1 = 2 × (Precision × Recall) / (Precision + Recall)`

F1 Score is useful when we want to give **equal importance to Precision and Recall**, especially when the classes are imbalanced.

A high F1 Score means that the model has a good balance between precision and recall.



## 16. Specificity

**Specificity**, also called the **True Negative Rate**, measures how well the model identifies actual negative cases.

Formula:

`Specificity = TN / (TN + FP)`

For example, if there are 100 actual negative cases and the model correctly identifies 90 of them, the specificity is 90%.

Specificity is particularly important when correctly identifying negative cases is important.



# ROC Curve and AUC

## 17. ROC Curve

**ROC** stands for **Receiver Operating Characteristic**.

The ROC Curve shows the performance of a binary classification model at different classification thresholds.

It plots:

* **True Positive Rate (TPR)** on the Y-axis
* **False Positive Rate (FPR)** on the X-axis

Recall is the same as True Positive Rate:

`TPR = TP / (TP + FN)`

False Positive Rate is:

`FPR = FP / (FP + TN)`

The ROC curve helps us understand how the model behaves when the classification threshold changes.



## 18. AUC

**AUC** stands for **Area Under the Curve**. It represents the area under the ROC curve.

AUC gives us a single number that summarizes how well the model can distinguish between the positive and negative classes.

Generally:

* **AUC = 1.0** → Excellent discrimination
* **AUC close to 1.0** → Very good model
* **AUC = 0.5** → No better than random guessing
* **AUC below 0.5** → Worse than random in the usual interpretation

A higher AUC generally means the model is better at ranking positive examples above negative examples.



## 19. Significance of ROC and AUC

ROC and AUC are useful because they evaluate model performance across **different classification thresholds**, rather than depending on only one threshold such as 0.5.

This is especially useful when we need to compare different classification models or when the appropriate threshold has not yet been selected.

For example, two models may have similar accuracy at a particular threshold, but their AUC values can show that one model has better overall ability to distinguish between the classes.



# Advantages of Logistic Regression

1. **Simple and easy to understand** – Logistic Regression is relatively easy to interpret compared with many complex machine learning algorithms.

2. **Good for classification** – It works well for binary classification and can also be extended to multiple classes.

3. **Provides probabilities** – Instead of only producing a class label, it can provide the probability of belonging to a class.

4. **Fast to train** – Logistic Regression is computationally efficient, especially for datasets with a reasonable number of features.

5. **Interpretable coefficients** – The coefficients can provide information about how the input features influence the prediction.

6. **Works well as a baseline model** – It is often a good first model to try before moving to more complex algorithms.



# Disadvantages of Logistic Regression

1. **Assumes a linear relationship between features and the log-odds** – Logistic Regression may not perform well when the decision boundary is highly nonlinear.

2. **Sensitive to multicollinearity** – Highly correlated input features can make coefficient estimates unstable and difficult to interpret.

3. **Can be affected by outliers** – Extreme observations can influence the model.

4. **May perform poorly on complex relationships** – Models such as decision trees, random forests, or neural networks can capture more complex patterns.

5. **Requires appropriate feature preparation** – Categorical variables usually need to be encoded, and feature scaling can be helpful depending on the optimization method and regularization.

6. **Can struggle with severe class imbalance** – Accuracy may become misleading, and additional techniques or metrics such as Precision, Recall, F1 Score, and AUC may be needed.
