# Unsupervised Learning, Clustering & Naive Bayes


# Unsupervised Learning

**Unsupervised Learning** is a type of machine learning where the model learns patterns and structures from data **without a predefined target/output label**.

In supervised learning, we have:

```text
Input → Target
```

In unsupervised learning, we generally have:

```text
Input Data → Discover Patterns
```

The algorithm tries to discover hidden structures, groups, or relationships in the data.


## Examples of Unsupervised Learning

Common unsupervised learning techniques include:

* K-Means Clustering
* Hierarchical Clustering
* DBSCAN
* Principal Component Analysis (PCA)
* Association Rule Learning


# Clustering

**Clustering** is an unsupervised learning technique used to divide data points into groups called **clusters**.

The goal is:

> Data points within the same cluster should be similar to each other, while data points in different clusters should be relatively different.

### Example

Suppose we have customer data:

* Age
* Income
* Spending behavior

A clustering algorithm might discover:

```text
Cluster 1 → Low income, low spending
Cluster 2 → High income, high spending
Cluster 3 → High income, low spending
```

We did not tell the algorithm these groups beforehand.

The algorithm discovered them from the data.

#  K-Means Clustering

**K-Means** is an unsupervised machine learning algorithm that divides data into a predefined number of clusters.

The `K` represents the number of clusters we want to create.

For example:

```text
K = 3
```

means we want the algorithm to create **3 clusters**.


# Goal of K-Means

The goal of K-Means is to create clusters where:

* Points within a cluster are close to each other.
* Points belonging to different clusters are relatively far apart.

K-Means does this by finding the best location for the **cluster centroids**.


# Centroid

A **centroid** is the center point of a cluster.

For example:

```text
        ●
    ●   ●   ●
        C
    ●       ●
```

`C` represents the centroid.

The centroid is calculated based on the mean position of the points assigned to that cluster.

# K-Means Algorithm

K-Means generally works through the following steps:

```text
1. Choose K
       ↓
2. Initialize K centroids
       ↓
3. Assign each data point to the nearest centroid
       ↓
4. Recalculate the centroids
       ↓
5. Reassign points
       ↓
6. Repeat
       ↓
7. Stop when assignments/centroids stabilize
```

# Step 1: Choose K

First, we decide how many clusters we want.

Example:

```text
K = 3
```

We want three clusters.

# Step 2: Initialize Centroids

The algorithm initially chooses `K` centroid positions.

```text
Centroid 1
Centroid 2
Centroid 3
```

The initial positions can affect the final result, which is why K-Means is often run with multiple initializations.


# Step 3: Assign Data Points

Each data point is assigned to the centroid that is closest to it.

Distance is commonly calculated using **Euclidean distance**.

```text
Distance =
√[(x₂ - x₁)² + (y₂ - y₁)²]
```

# Step 4: Recalculate Centroids

After assigning the points to clusters, the algorithm calculates a new centroid for each cluster.

The new centroid is based on the mean of the points belonging to that cluster.


# Step 5: Repeat

The algorithm repeatedly:

```text
Assign points
     ↓
Calculate centroids
     ↓
Reassign points
     ↓
Calculate centroids
```

until the clusters stabilize or the algorithm reaches its stopping criteria.


# K-Means Objective

K-Means tries to minimize the **within-cluster sum of squared distances**.

This is commonly called **inertia** or **within-cluster sum of squares (WCSS)**.

Conceptually:

```text
WCSS = Σ distance(point, centroid)²
```

Lower WCSS means points are, on average, closer to their cluster centroids.

# Elbow Method

The **Elbow Method** is used to help choose an appropriate value of `K` for K-Means.

We run K-Means with different values of K and calculate the WCSS/inertia for each K.

Example:

```text
K = 1 → High WCSS
K = 2 → Lower WCSS
K = 3 → Lower WCSS
K = 4 → Much lower WCSS
K = 5 → Slight improvement
K = 6 → Very small improvement
```

We plot:

```text
K → X-axis
WCSS → Y-axis
```

# Why Is It Called the Elbow Method?

As K increases, WCSS generally decreases.

At some point, increasing K provides only a small improvement.

The point where the improvement begins to slow down is called the **elbow**.

```text
WCSS
  |
  |\
  | \
  |  \
  |   \
  |    \__
  |       \___
  |
  +---------------- K
        ↑
      Elbow
```

The elbow can be considered a reasonable candidate for the number of clusters.

# Silhouette Score

The **Silhouette Score** measures how well a data point fits within its assigned cluster compared with other clusters.

It considers:

1. How close a point is to other points in its own cluster.
2. How close the point is to points in the nearest other cluster.


The silhouette score asks:

> "Is this point more similar to its own cluster than to another cluster?"


# Silhouette Score Range

The score ranges from:

```text
-1 to +1
```

### Score close to +1

The point is well matched to its own cluster.

```text
Good clustering
```

### Score around 0

The point may be located near the boundary between clusters.

```text
Unclear cluster assignment
```

### Negative score

The point may be closer to another cluster than its assigned cluster.

```text
Potentially incorrect assignment
```

# Silhouette Score Formula

For a data point:

```text
s = (b - a) / max(a, b)
```

Where:

* `a` = average distance from the point to other points in its own cluster
* `b` = average distance from the point to points in the nearest other cluster


# K-Means vs Silhouette Score

K-Means creates the clusters.

The Silhouette Score helps us evaluate **how well those clusters are separated and formed**.

```text
K-Means
   ↓
Create clusters
   ↓
Silhouette Score
   ↓
Evaluate cluster quality
```


# Hierarchical Clustering

Hierarchical Clustering is an unsupervised learning technique that creates a hierarchy of clusters.

Instead of directly choosing a fixed number of clusters, it builds a tree-like structure showing how data points or groups are merged or separated.


# Types of Hierarchical Clustering

## 1. Agglomerative Clustering

This is a **bottom-up approach**.

Initially:

```text
Every data point = Separate cluster
```

Then the closest clusters are repeatedly merged.

```text
A   B   C   D

↓ Merge

AB   C   D

↓ Merge

AB   CD

↓ Merge

ABCD
```

## 2. Divisive Clustering

This is a **top-down approach**.

Initially:

```text
All data points = One cluster
```

Then the cluster is repeatedly split into smaller clusters.


# Dendrogram

A **dendrogram** is a tree-like diagram used to visualize hierarchical clustering.

It shows:

* Which points/clusters were joined
* The order in which they were joined
* The distance at which they were joined

Example:

```text
        ┌───────────────┐
        │               │
      ┌─┴─┐           ┌─┴─┐
      A   B             C   D
```

The horizontal/vertical structure depends on the plotting convention, but the important idea is that the dendrogram represents the hierarchy of merges.


# Choosing Clusters from a Dendrogram

We can draw a horizontal line across the dendrogram.

The number of branches intersected by that line gives the number of clusters.

```text
        ┌───────────────┐
        │               │
      ┌─┴─┐           ┌─┴─┐
      A   B             C   D
      ───────────────────────
```

Changing the cutting level can produce a different number of clusters.


# DBSCAN

DBSCAN stands for:

**Density-Based Spatial Clustering of Applications with Noise**

DBSCAN is a density-based clustering algorithm.

Unlike K-Means, DBSCAN does not require us to specify the number of clusters beforehand.

It groups together points that are located in **dense regions** and identifies isolated points as noise.


#  Main Idea of DBSCAN

DBSCAN looks at the density of data points.

```text
Dense region → Cluster

Sparse/isolated region → Potential noise
```

This makes DBSCAN useful when clusters have irregular shapes.


# Important DBSCAN Terms

DBSCAN primarily uses two parameters:

* `ε (epsilon)`
* `min_samples` / `min_points`

These parameters determine whether a region contains enough nearby points to be considered dense.


# Epsilon (ε)

**Epsilon** defines the radius around a data point that DBSCAN uses to search for neighboring points.

In simple terms:

> Epsilon determines how far we look around a point for neighbors.

```text
        ●
     ●  ●  ●
        P
     ●     ●

<--- ε --->
```

A larger epsilon means a larger neighborhood.


#  Min Points

`min_points` specifies the minimum number of points that must exist within the epsilon neighborhood for a point to be considered a **core point**.

For example:

```text
epsilon = 0.5
min_points = 5
```

means the algorithm looks within a radius of 0.5 and requires enough nearby points to satisfy the density requirement.

#  Core Point

A **core point** has at least the required number of neighboring points within the epsilon radius.

Example:

```text
        ●
    ●   ●   ●
      ● P ●
        ●
```

If enough points are inside the epsilon neighborhood:

```text
P = Core Point
```

# Border Point

A **border point** does not have enough neighbors to be a core point itself, but it is located within the epsilon neighborhood of a core point.

```text
Core Point
    ● ● ●
   ● ● ●
       ○
       ↑
   Border Point
```

The border point belongs to the cluster because it is reachable from a core point.


#  Noise Point

A **noise point** is a point that:

* Is not a core point
* Is not sufficiently connected to a core point

It is treated as an outlier/noise.

```text
● ● ●        ● ●

   Cluster       ●
                 ↑
              Noise
```


#  DBSCAN Process

```text
Choose ε and min_points
          ↓
Examine each point
          ↓
Find neighboring points
          ↓
Identify core points
          ↓
Expand clusters from core points
          ↓
Include reachable border points
          ↓
Mark remaining points as noise
```

# K-Means vs Hierarchical vs DBSCAN

| K-Means                            | Hierarchical                          | DBSCAN                                      |
| ---------------------------------- | ------------------------------------- | ------------------------------------------- |
| Requires K                         | Does not require K initially          | Does not require K                          |
| Centroid-based                     | Tree/hierarchy-based                  | Density-based                               |
| Uses centroids                     | Uses distances/linkage                | Uses density                                |
| Sensitive to outliers              | Can be affected by outliers           | Can identify noise                          |
| Works well with compact clusters   | Useful for hierarchical relationships | Works well with irregularly shaped clusters |
| Uses Elbow Method to help choose K | Uses dendrogram                       | Uses epsilon and min_points                 |


# When to Use Which Clustering Algorithm?

## Use K-Means when:

* You know approximately how many groups you want
* Clusters are relatively compact
* Data is suitable for centroid-based clustering
* You need a simple and scalable clustering method


## Use Hierarchical Clustering when:

* You want to understand relationships between clusters
* You want a dendrogram
* You do not know the final number of clusters initially
* Dataset size is manageable


## Use DBSCAN when:

* Clusters may have irregular shapes
* You have noisy or outlier data
* You want to identify noise
* Density-based grouping makes sense


# Naive Bayes

**Naive Bayes** is a **supervised machine learning algorithm** primarily used for classification.

It is based on **Bayes' Theorem**.

Despite the name "Naive," it can be very effective, especially for text classification.


# Bayes' Theorem

The basic formula is:

```text
P(A|B) = [P(B|A) × P(A)] / P(B)
```

Where:

* `P(A|B)` = Probability of A given B
* `P(B|A)` = Probability of B given A
* `P(A)` = Prior probability of A
* `P(B)` = Probability of B


# Why Is It Called "Naive"?

Naive Bayes makes a strong assumption that the features are **conditionally independent given the class**.

For example, in spam detection:

```text
Email contains "free"
Email contains "offer"
Email contains "winner"
```

Naive Bayes treats these features as conditionally independent given whether the email is spam, even though real-world words can be related.

This assumption simplifies the calculation.

# Example of Naive Bayes

Suppose we want to classify emails as:

```text
Spam
Not Spam
```

The email contains:

```text
"Win free money"
```

The model calculates the probability that the email belongs to each class.

```text
P(Spam | words)
P(Not Spam | words)
```

The class with the higher probability is selected.



#  Naive Bayes Workflow

```text
Training Data
      ↓
Calculate class probabilities
      ↓
Calculate feature probabilities
      ↓
Receive new data
      ↓
Calculate probability for each class
      ↓
Choose class with highest probability
```


# Common Applications of Naive Bayes

Naive Bayes is commonly used for:

* Spam detection
* Sentiment analysis
* Text classification
* Document classification
* News classification
* Topic classification


# Types of Naive Bayes

## 1. Gaussian Naive Bayes

Used mainly when numerical features can reasonably be modeled using a Gaussian/normal distribution.

Example:

```text
Age
Height
Weight
```

## 2. Multinomial Naive Bayes

Commonly used for text classification where features represent counts or frequencies.

Example:

```text
Word counts
Term frequencies
```

## 3. Bernoulli Naive Bayes

Used when features are binary.

Example:

```text
Word appears → 1
Word does not appear → 0
```

# Unsupervised vs Supervised Learning

It is important to remember that **clustering and Naive Bayes are different types of machine learning**.

| Unsupervised Learning   | Supervised Learning          |

| Finds hidden patterns   | Learns from labeled examples |
| K-Means                 | Linear Regression            |
| Hierarchical Clustering | Logistic Regression          |
| DBSCAN                  | Naive Bayes                  |
| PCA                     | Decision Tree                |
